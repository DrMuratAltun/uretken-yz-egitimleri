## 1. Erişim

notebooklm.google.com → Google hesabınla giriş. Türkiye'den direkt erişim var, ücretsiz.


## 2. Kaynak Türleri

50 kaynak/notebook (Free), Plus'ta daha fazla. Tipler: PDF, Google Docs/Slides, web URL, YouTube linki, paste markdown.


## 3. İlk Notebook

New notebook → 5-10 PDF yükle → "Generate" tuşuna basma, soru sormaya başla.


## 4. Sınırlar

500 MB / kaynak, kombineli ~25M kelime. YouTube transcript otomatik. Web URL sayfasının metnini alır (görsel anlamaz).
