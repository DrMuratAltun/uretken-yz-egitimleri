## 3. Betimsel İstatistik

### 3.1 Merkezi Eğilim Ölçüleri

```python
# total_bill (toplam hesap) sütunu üzerinden betimsel istatistik
hesap = df['total_bill']

print("=" * 50)
print("TOPLAM HESAP — BETİMSEL İSTATİSTİK")
print("=" * 50)
print(f"Ortalama (Mean):         {np.mean(hesap):.2f}")
print(f"Medyan (Median):         {np.median(hesap):.2f}")
print(f"Mod (Mode):              {stats.mode(hesap, keepdims=True).mode[0]:.2f}")
print(f"Standart Sapma (Std):    {np.std(hesap, ddof=1):.2f}")
print(f"Varyans (Variance):      {np.var(hesap, ddof=1):.2f}")
print(f"Minimum:                 {hesap.min():.2f}")
print(f"Maksimum:                {hesap.max():.2f}")
print(f"Çeyrekler Arası Açıklık: {hesap.quantile(0.75) - hesap.quantile(0.25):.2f}")
```

### 3.2 Bahşiş Sütunu için Betimsel İstatistik

```python
bahsis = df['tip']

print("=" * 50)
print("BAHŞİŞ — BETİMSEL İSTATİSTİK")
print("=" * 50)
print(f"Ortalama:          {bahsis.mean():.2f}")
print(f"Medyan:            {bahsis.median():.2f}")
print(f"Standart Sapma:    {bahsis.std():.2f}")
print(f"Varyans:           {bahsis.var():.2f}")
print(f"Çarpıklık (Skew):  {bahsis.skew():.2f}")
print(f"Basıklık (Kurt):   {bahsis.kurtosis():.2f}")
```

## 4. Normal Dağılım ve Histogram

Normal dağılım, istatistikte en önemli dağılımdır. Bir değişkenin normal dağılıp dağılmadığını görsel olarak inceleyeceğiz.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Toplam Hesap ---
ax = axes[0]
ax.hist(hesap, bins=25, density=True, alpha=0.7, color='steelblue', edgecolor='white', label='Veri')

# Normal eğri çizimi
mu, sigma = hesap.mean(), hesap.std()
x = np.linspace(hesap.min(), hesap.max(), 200)
ax.plot(x, stats.norm.pdf(x, mu, sigma), 'r-', linewidth=2, label=f'Normal Eğri\n(μ={mu:.1f}, σ={sigma:.1f})')
ax.set_title('Toplam Hesap Dağılımı')
ax.set_xlabel('Toplam Hesap ($)')
ax.set_ylabel('Yoğunluk')
ax.legend()

# --- Bahşiş ---
ax = axes[1]
ax.hist(bahsis, bins=25, density=True, alpha=0.7, color='coral', edgecolor='white', label='Veri')

mu2, sigma2 = bahsis.mean(), bahsis.std()
x2 = np.linspace(bahsis.min(), bahsis.max(), 200)
ax.plot(x2, stats.norm.pdf(x2, mu2, sigma2), 'r-', linewidth=2, label=f'Normal Eğri\n(μ={mu2:.1f}, σ={sigma2:.1f})')
ax.set_title('Bahşiş Dağılımı')
ax.set_xlabel('Bahşiş ($)')
ax.set_ylabel('Yoğunluk')
ax.legend()

plt.tight_layout()
plt.show()
```

## 5. Normallik Testi — Shapiro-Wilk

**Shapiro-Wilk Testi:**
- H₀ (Sıfır Hipotez): Veri normal dağılmaktadır.
- H₁ (Alternatif Hipotez): Veri normal dağılmamaktadır.
- p < 0.05 → H₀ reddedilir → Veri normal dağılmıyor.
- p ≥ 0.05 → H₀ reddedilemez → Veri normal dağılıma uygun olabilir.

```python
# Toplam Hesap için Shapiro-Wilk testi
stat_hesap, p_hesap = stats.shapiro(hesap)
print("TOPLAM HESAP — Shapiro-Wilk Testi")
print(f"  Test İstatistiği: {stat_hesap:.4f}")
print(f"  p-değeri:         {p_hesap:.4f}")
print(f"  Sonuç:            {'Normal dağılmıyor (H₀ RED)' if p_hesap < 0.05 else 'Normal dağılıma uygun (H₀ kabul)'}")
print()

# Bahşiş için Shapiro-Wilk testi
stat_bahsis, p_bahsis = stats.shapiro(bahsis)
print("BAHŞİŞ — Shapiro-Wilk Testi")
print(f"  Test İstatistiği: {stat_bahsis:.4f}")
print(f"  p-değeri:         {p_bahsis:.4f}")
print(f"  Sonuç:            {'Normal dağılmıyor (H₀ RED)' if p_bahsis < 0.05 else 'Normal dağılıma uygun (H₀ kabul)'}")
```

## 6. Korelasyon Analizi

İki değişken arasındaki ilişkiyi ölçmek için korelasyon kullanılır.

| Korelasyon | Aralık | Yorum |
|---|---|---|
| r ≈ 1 | Güçlü pozitif | Biri artarken diğeri de artar |
| r ≈ -1 | Güçlü negatif | Biri artarken diğeri azalır |
| r ≈ 0 | Zayıf/Yok | Doğrusal ilişki yok |

### 6.1 Pearson Korelasyonu (Doğrusal ilişki)

```python
# Pearson Korelasyonu: total_bill vs tip
r_pearson, p_pearson = stats.pearsonr(df['total_bill'], df['tip'])
print(f"Pearson Korelasyon Katsayısı (r): {r_pearson:.4f}")
print(f"p-değeri:                         {p_pearson:.6f}")
print(f"Yorum: {'İstatistiksel olarak anlamlı' if p_pearson < 0.05 else 'Anlamlı değil'} (α=0.05)")
```

### 6.2 Spearman Korelasyonu (Sıralama tabanlı ilişki)

```python
# Spearman Korelasyonu: total_bill vs tip
r_spearman, p_spearman = stats.spearmanr(df['total_bill'], df['tip'])
print(f"Spearman Korelasyon Katsayısı (ρ): {r_spearman:.4f}")
print(f"p-değeri:                          {p_spearman:.6f}")
print(f"Yorum: {'İstatistiksel olarak anlamlı' if p_spearman < 0.05 else 'Anlamlı değil'} (α=0.05)")
```

### 6.3 Saçılım Grafiği ile Korelasyon Görselleştirme

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', style='sex',
                palette='Set1', s=80, alpha=0.7, ax=ax)

# Regresyon doğrusu
z = np.polyfit(df['total_bill'], df['tip'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['total_bill'].min(), df['total_bill'].max(), 100)
ax.plot(x_line, p(x_line), 'r--', linewidth=2, label=f'Regresyon (r={r_pearson:.3f})')

ax.set_title('Toplam Hesap ve Bahşiş İlişkisi')
ax.set_xlabel('Toplam Hesap ($)')
ax.set_ylabel('Bahşiş ($)')
ax.legend()
plt.tight_layout()
plt.show()
```

### 6.4 Korelasyon Matrisi

### Korelasyon Analizi ve Isı Haritası

Özellikler arasındaki ilişkiyi korelasyon matrisi ve ısı haritası ile görselleştiriyoruz. Yüksek korelasyonlu özellikler modelin en çok yararlanacağı özelliklerdir.

```python
# Sayısal sütunlar için korelasyon matrisi
korelasyon = df[['total_bill', 'tip', 'size']].corr()

fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(korelasyon, annot=True, fmt='.3f', cmap='coolwarm', center=0,
            square=True, linewidths=1, ax=ax)
ax.set_title('Korelasyon Matrisi (Pearson)')
plt.tight_layout()
plt.show()
```

## 7. Güven Aralığı Hesaplama

Güven aralığı, bir popülasyon parametresinin (ör. ortalama) belirli bir olasılıkla bulunacağı değer aralığıdır.

**%95 Güven Aralığı Formülü:**

$$\bar{x} \pm t_{\alpha/2} \cdot \frac{s}{\sqrt{n}}$$

```python
def guven_araligi(veri, guven_duzeyi=0.95):
    """Verilen veri için güven aralığı hesaplar."""
    n = len(veri)
    ortalama = np.mean(veri)
    se = stats.sem(veri)  # Standart hata
    h = se * stats.t.ppf((1 + guven_duzeyi) / 2, n - 1)
    return ortalama, ortalama - h, ortalama + h

# Toplam Hesap için %95 güven aralığı
ort, alt, ust = guven_araligi(df['total_bill'])
print("TOPLAM HESAP — %95 Güven Aralığı")
print(f"  Ortalama:   {ort:.2f}")
print(f"  Alt Sınır:  {alt:.2f}")
print(f"  Üst Sınır:  {ust:.2f}")
print(f"  Aralık:     [{alt:.2f}, {ust:.2f}]")
print()

# Bahşiş için %95 güven aralığı
ort2, alt2, ust2 = guven_araligi(df['tip'])
print("BAHŞİŞ — %95 Güven Aralığı")
print(f"  Ortalama:   {ort2:.2f}")
print(f"  Alt Sınır:  {alt2:.2f}")
print(f"  Üst Sınır:  {ust2:.2f}")
print(f"  Aralık:     [{alt2:.2f}, {ust2:.2f}]")
```

### Farklı güven düzeyleri ile karşılaştırma

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Farklı güven düzeyleri ile karşılaştırma
print("TOPLAM HESAP — Farklı Güven Düzeyleri")
print("-" * 45)
for duzey in [0.90, 0.95, 0.99]:
    ort, alt, ust = guven_araligi(df['total_bill'], duzey)
    print(f"  %{int(duzey*100)} Güven Aralığı: [{alt:.2f}, {ust:.2f}]  (Genişlik: {ust-alt:.2f})")
```
