> **Hafta 4 · NotebookLM ve Bilgi Yönetimi** · *Üretken YZ Atölyesi · Dr. Murat Altun*

**Bu notebook hakkında:** NotebookLM'in en sevilen özelliği — kaynaklarından otomatik podcast üretimi.

---


## 1. Audio Overview Nedir?

2 AI sunucu, kaynaklarından bir podcast yapar. Doğal sohbet, sorular, espriler.


## 2. Türkçe Destek

"Customize" → dil → Türkçe seç. Hâlâ İngilizce karışık olabilir; sistem mesajıyla zorla Türkçe.


## 3. Customize Talimatları

Hedef kitle, ton, vurgu yapılacak konular — sen belirle.


## 4. İndir ve Paylaş

MP3 olarak indir, Spotify/YouTube'a yükle, ya da Google Drive ile paylaş.
