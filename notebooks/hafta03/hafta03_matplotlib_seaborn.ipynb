## 1. Kütüphanelerin Yüklenmesi

### Kütüphanelerin Yüklenmesi

Projede kullanacağımız kütüphaneleri içe aktarıyoruz:

| Kütüphane | Amacı |
|-----------|-------|
| `matplotlib` | Grafik ve görselleştirme |
| `numpy` | Sayısal hesaplamalar ve dizi işlemleri |
| `pandas` | Veri çerçeveleri (DataFrame) ile veri analizi |
| `seaborn` | İstatistiksel görselleştirme (Matplotlib üzerine kurulu) |


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Grafiklerin notebook içinde görünmesi için
%matplotlib inline

# Seaborn stilini ayarlayalım
sns.set_style("whitegrid")

print("Kütüphaneler başarıyla yüklendi!")
```

## 2. Matplotlib Temel Grafikler

### 2.1 Çizgi Grafiği (Line Plot)

```python
# Basit çizgi grafiği
aylar = ["Oca", "Şub", "Mar", "Nis", "May", "Haz", "Tem", "Ağu", "Eyl", "Eki", "Kas", "Ara"]
sicaklik = [5, 7, 12, 17, 22, 27, 30, 29, 25, 18, 12, 7]

plt.figure(figsize=(10, 5))
plt.plot(aylar, sicaklik, marker='o', color='tomato', linewidth=2, markersize=8)
plt.title("Antalya Aylık Ortalama Sıcaklık", fontsize=16, fontweight='bold')
plt.xlabel("Aylar", fontsize=12)
plt.ylabel("Sıcaklık (°C)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 2.2 Çubuk Grafiği (Bar Plot)

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Çubuk grafiği
diller = ["Python", "JavaScript", "Java", "C++", "TypeScript", "Go"]
kullanici_sayisi = [15000, 12000, 9000, 7000, 8500, 5000]
renkler = ["#3776ab", "#f7df1e", "#f89820", "#00599c", "#3178c6", "#00add8"]

plt.figure(figsize=(10, 6))
bars = plt.bar(diller, kullanici_sayisi, color=renkler, edgecolor='black', linewidth=0.5)
plt.title("Programlama Dilleri Kullanıcı Sayısı", fontsize=16, fontweight='bold')
plt.xlabel("Programlama Dili", fontsize=12)
plt.ylabel("Kullanıcı Sayısı", fontsize=12)

# Çubukların üzerine değer yazalım
for bar, val in zip(bars, kullanici_sayisi):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 200,
             f"{val:,}", ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

### 2.3 Saçılım Grafiği (Scatter Plot)

### Saçılım Grafiği

İki değişken arasındaki ilişkiyi saçılım grafiği ile inceliyoruz. Noktaların oluşturduğu desen, doğrusal veya doğrusal olmayan ilişkiyi gösterir.

```python
# Saçılım grafiği
np.random.seed(42)
boy = np.random.normal(170, 10, 100)
kilo = boy * 0.5 + np.random.normal(0, 5, 100) - 10

plt.figure(figsize=(8, 6))
plt.scatter(boy, kilo, c=kilo, cmap='coolwarm', alpha=0.7, s=60, edgecolors='black', linewidth=0.5)
plt.colorbar(label='Kilo (kg)')
plt.title("Boy - Kilo İlişkisi", fontsize=16, fontweight='bold')
plt.xlabel("Boy (cm)", fontsize=12)
plt.ylabel("Kilo (kg)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 2.4 Histogram

### Dağılım Görselleştirmesi

Aşağıdaki grafikte verinin dağılımını histogram ile inceliyoruz. Dağılımın şekli (normal, çarpık, bimodal) hangi istatistiksel yöntemlerin uygulanabileceğini belirler.

```python
# Histogram
np.random.seed(42)
notlar = np.random.normal(70, 15, 500)
notlar = np.clip(notlar, 0, 100)  # 0-100 arasında sınırla

plt.figure(figsize=(10, 6))
plt.hist(notlar, bins=20, color='steelblue', edgecolor='black', alpha=0.7)
plt.axvline(notlar.mean(), color='red', linestyle='--', linewidth=2, label=f'Ortalama: {notlar.mean():.1f}')
plt.axvline(np.median(notlar), color='green', linestyle='--', linewidth=2, label=f'Medyan: {np.median(notlar):.1f}')
plt.title("Öğrenci Not Dağılımı", fontsize=16, fontweight='bold')
plt.xlabel("Not", fontsize=12)
plt.ylabel("Öğrenci Sayısı", fontsize=12)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

### 2.5 Pasta Grafiği (Pie Chart)

### Pasta Grafiği

Kategorilerin genel dağılım içindeki oranlarını pasta grafiği ile görselleştiriyoruz.

```python
# Pasta grafiği
kategoriler = ["Gıda", "Ulaşım", "Kira", "Eğlence", "Eğitim", "Diğer"]
harcamalar = [2500, 1200, 4000, 800, 1500, 1000]
patlama = [0, 0, 0.1, 0, 0, 0]  # Kira dilimini öne çıkar
renkler = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff66b3', '#c2c2f0']

plt.figure(figsize=(8, 8))
plt.pie(harcamalar, labels=kategoriler, autopct='%1.1f%%', explode=patlama,
        colors=renkler, shadow=True, startangle=90, textprops={'fontsize': 12})
plt.title("Aylık Harcama Dağılımı", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Grafik Özelleştirme

Matplotlib'de grafikleri özelleştirmek için birçok parametre kullanabiliriz.

```python
# Birden fazla çizgi ve özelleştirme
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure(figsize=(12, 6))
plt.plot(x, y1, 'r-', linewidth=2, label='sin(x)')
plt.plot(x, y2, 'b--', linewidth=2, label='cos(x)')
plt.plot(x, y3, 'g-.', linewidth=2, label='sin(x)·cos(x)')

plt.title("Trigonometrik Fonksiyonlar", fontsize=16, fontweight='bold')
plt.xlabel("x değeri", fontsize=12)
plt.ylabel("y değeri", fontsize=12)
plt.legend(loc='upper right', fontsize=11, framealpha=0.9)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.axvline(x=0, color='black', linewidth=0.5)
plt.tight_layout()
plt.show()
```

## 4. Alt Grafikler (Subplots)

`plt.subplots()` ile tek bir figürde birden fazla grafik oluşturabiliriz.

```python
# 2x2 alt grafik düzeni
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Sol üst: Çizgi grafiği
x = np.linspace(0, 10, 50)
axes[0, 0].plot(x, np.sin(x), 'r-o', markersize=3)
axes[0, 0].set_title("Çizgi Grafiği", fontsize=13, fontweight='bold')
axes[0, 0].set_xlabel("x")
axes[0, 0].set_ylabel("sin(x)")
axes[0, 0].grid(True, alpha=0.3)

# Sağ üst: Çubuk grafiği
kategoriler = ["A", "B", "C", "D", "E"]
degerler = [23, 45, 56, 78, 32]
axes[0, 1].bar(kategoriler, degerler, color='steelblue', edgecolor='black')
axes[0, 1].set_title("Çubuk Grafiği", fontsize=13, fontweight='bold')

# Sol alt: Histogram
veri = np.random.normal(50, 10, 1000)
axes[1, 0].hist(veri, bins=30, color='coral', edgecolor='black', alpha=0.7)
axes[1, 0].set_title("Histogram", fontsize=13, fontweight='bold')

# Sağ alt: Saçılım grafiği
x_sc = np.random.rand(50)
y_sc = x_sc * 2 + np.random.normal(0, 0.3, 50)
axes[1, 1].scatter(x_sc, y_sc, c='green', alpha=0.6, edgecolors='black')
axes[1, 1].set_title("Saçılım Grafiği", fontsize=13, fontweight='bold')

fig.suptitle("Matplotlib Alt Grafik Örnekleri", fontsize=18, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 5. Seaborn ile Görselleştirme

Seaborn, Matplotlib üzerine kurulu daha yüksek seviyeli bir görselleştirme kütüphanesidir.

### 5.1 Veri Setlerini Yükleme

```python
# Seaborn yerleşik veri setleri
tips = sns.load_dataset("tips")
titanic = sns.load_dataset("titanic")

print("Tips veri seti:")
print(tips.head())
print(f"\nBoyut: {tips.shape}")
print("\n" + "="*50)
print("\nTitanic veri seti:")
print(titanic.head())
print(f"\nBoyut: {titanic.shape}")
```

### 5.2 Sayma Grafiği (Count Plot)

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Titanic - Sınıfa göre yolcu sayısı
plt.figure(figsize=(10, 6))
sns.countplot(data=titanic, x='class', hue='survived', palette='Set2')
plt.title("Titanic: Sınıfa Göre Hayatta Kalma Durumu", fontsize=16, fontweight='bold')
plt.xlabel("Yolcu Sınıfı", fontsize=12)
plt.ylabel("Yolcu Sayısı", fontsize=12)
plt.legend(title='Hayatta Kaldı', labels=['Hayır', 'Evet'])
plt.tight_layout()
plt.show()
```

### 5.3 Kutu Grafiği (Box Plot)

### Kutu Grafiği (Box Plot)

Kutu grafiği verinin dağılımını, medyanını, çeyrekliklerini ve uç değerlerini (outlier) gösterir. Gruplar arası karşılaştırma için idealdir.

```python
# Tips - Güne göre bahşiş dağılımı
plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='tip', hue='sex', palette='pastel')
plt.title("Güne ve Cinsiyete Göre Bahşiş Dağılımı", fontsize=16, fontweight='bold')
plt.xlabel("Gün", fontsize=12)
plt.ylabel("Bahşiş ($)", fontsize=12)
plt.legend(title='Cinsiyet')
plt.tight_layout()
plt.show()
```

### 5.4 Dağılım Grafiği (Histplot)

### Dağılım Görselleştirmesi

Aşağıdaki grafikte verinin dağılımını histogram ile inceliyoruz. Dağılımın şekli (normal, çarpık, bimodal) hangi istatistiksel yöntemlerin uygulanabileceğini belirler.

```python
# Titanic - Yaş dağılımı
plt.figure(figsize=(10, 6))
sns.histplot(data=titanic, x='age', hue='survived', kde=True, bins=30, palette='Set1', alpha=0.6)
plt.title("Titanic: Yaşa Göre Hayatta Kalma Dağılımı", fontsize=16, fontweight='bold')
plt.xlabel("Yaş", fontsize=12)
plt.ylabel("Yolcu Sayısı", fontsize=12)
plt.legend(title='Hayatta Kaldı', labels=['Hayır', 'Evet'])
plt.tight_layout()
plt.show()
```

### 5.5 Isı Haritası (Heatmap) - Korelasyon Matrisi

```python
# Tips veri seti korelasyon matrisi
sayisal_sutunlar = tips.select_dtypes(include=[np.number])
korelasyon = sayisal_sutunlar.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(korelasyon, annot=True, cmap='coolwarm', center=0,
            fmt='.2f', linewidths=1, linecolor='white',
            square=True, cbar_kws={'shrink': 0.8})
plt.title("Tips Veri Seti - Korelasyon Matrisi", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Korelasyon Analizi ve Isı Haritası

Özellikler arasındaki ilişkiyi korelasyon matrisi ve ısı haritası ile görselleştiriyoruz. Yüksek korelasyonlu özellikler modelin en çok yararlanacağı özelliklerdir.

```python
# Titanic korelasyon matrisi
titanic_sayisal = titanic.select_dtypes(include=[np.number])
korelasyon_titanic = titanic_sayisal.corr()

plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(korelasyon_titanic, dtype=bool))  # Üst üçgeni maskele
sns.heatmap(korelasyon_titanic, annot=True, cmap='RdYlBu_r', center=0,
            fmt='.2f', linewidths=1, mask=mask, square=True)
plt.title("Titanic Veri Seti - Korelasyon Matrisi", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5.6 Çift Değişken Grafiği (Pair Plot)

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Tips veri seti ile pairplot
sns.pairplot(tips, hue='sex', palette='Set1', diag_kind='kde', height=2.5)
plt.suptitle("Tips Veri Seti - Çift Değişken Grafiği", fontsize=16, fontweight='bold', y=1.02)
plt.show()
```
