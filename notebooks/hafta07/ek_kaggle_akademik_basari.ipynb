```python
import pandas as pd
from pycaret import classification
```

## Veri Setlerini Yükle ##

### Veri Setinin Yüklenmesi

Veri setini yükleyip temel bilgilerine bakıyoruz.

```python
train = pd.read_csv("/kaggle/input/playground-series-s4e6/train.csv")
test = pd.read_csv("/kaggle/input/playground-series-s4e6/test.csv")
sub = pd.read_csv("/kaggle/input/playground-series-s4e6/sample_submission.csv")
```

## Model Kur (Setup) ##

```python
clf1 = classification.setup(data = train, 
             target = 'Target',
             normalize = True
             )
```

## Modelleri Karşılaştır: En iyi model ##

```python
classification.compare_models()
```

## En İyi Model Cross Validation ##

```python
classification_lgbm = classification.create_model('lightgbm')
```

## En İyi Modeli İçin İnce Ayar (Fine Tuning) ##

```python
tuned_lgbm_model = classification.tune_model(classification_lgbm)
```

## En İyi Modelin Parametreleri ##

```python
print(tuned_lgbm_model)
```

## Performnas ROC Curve ## 

```python
classification.plot_model(tuned_lgbm_model)
```

## Öğrenme Eğrisi ##

```python
classification.plot_model(estimator = tuned_lgbm_model, plot = 'learning')
```

## Hata Matrisi ##

```python
classification.plot_model(tuned_lgbm_model, plot = 'confusion_matrix')
```

```python
classification.plot_model(tuned_lgbm_model, plot = 'error')
```

## Özelliklerin Önemleri ##

```python
classification.plot_model(tuned_lgbm_model, plot='feature')
```

## Modelin Değerlendirilmesi ##

```python
classification.evaluate_model(tuned_lgbm_model)
```

## Modeli Kaydet ##

### Kütüphanelerin Yüklenmesi

| Kütüphane | Amacı |

| `pickle` | Nesne kaydetme |


```python
import pickle
with open('tuned_lgbm_model.pkl', 'wb') as f:
    pickle.dump(tuned_lgbm_model, f)
```
