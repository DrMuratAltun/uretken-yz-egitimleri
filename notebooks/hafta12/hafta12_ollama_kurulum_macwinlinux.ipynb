İlk soru: "Merhaba, kendini Türkçe tanıt."


## Python Client




````python
import ollama



# İlk konuşma

response = ollama.chat(model="llama3.2:3b", messages=[

    {"role": "user", "content": "Türkiye'nin başkenti neresi? Tek cümleyle cevap ver."}

])



print(response["message"]["content"])

````

## Türkçe Test




````python
# Aynı soruyu 3 model için karşılaştır

modeller = ["llama3.2:3b", "qwen2.5:7b", "gemma2:9b"]

soru = "Türkçe bir atasözü açıkla: 'Damlaya damlaya göl olur.'"



for m in modeller:

    try:

        r = ollama.chat(model=m, messages=[{"role": "user", "content": soru}])

        print(f"--- {m} ---")

        print(r["message"]["content"][:200])

        print()

    except Exception as e:

        print(f"{m}: {e}")

````

> İlk çalıştırmada model indirilir (1-5 dk). Bir kez indirildikten sonra çevrimdışı.


## Streaming (Akış Halinde Yanıt)




````python
# ChatGPT gibi yazarken görmek

stream = ollama.chat(

    model="llama3.2:3b",

    messages=[{"role": "user", "content": "Üretken YZ'yi 3 paragrafla anlat."}],

    stream=True

)



for chunk in stream:

    print(chunk["message"]["content"], end="", flush=True)

````
