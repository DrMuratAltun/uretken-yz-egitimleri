## Test Görevleri

10 Türkçe görev:

## Otomatik Kıyaslama




```python
import ollama

import pandas as pd

from datetime import datetime



gorevler = [

    "Türkiye'nin başkenti hangi şehirdir?",

    "Damlaya damlaya göl olur deyimini açıkla.",

    "5 yaşındaki bir çocuğa atomu Türkçe anlat.",

    "Bir öğretmen için 5E modelinde 1 paragraf ders planı yaz.",

    "Karikatür önerisi: öğretmen ve üretken YZ.",

]



modeller = ["llama3.2:3b", "qwen2.5:7b", "gemma2:9b"]

sonuclar = []



for g in gorevler:

    for m in modeller:

        baslangic = datetime.now()

        try:

            r = ollama.chat(model=m, messages=[{"role": "user", "content": g}])

            cevap = r["message"]["content"][:300]

            sure = (datetime.now() - baslangic).total_seconds()

            sonuclar.append({"gorev": g[:50], "model": m, "cevap": cevap, "saniye": round(sure, 1)})

        except Exception as e:

            sonuclar.append({"gorev": g[:50], "model": m, "cevap": f"HATA: {e}", "saniye": -1})



df = pd.DataFrame(sonuclar)

df.to_csv("kiyas_sonuclari.csv", index=False)

df

```
