## System Message




```python
from ollama import chat



cevap = chat(

    model="llama3.2:3b",

    messages=[

        {"role": "system", "content": "Sen 20 yıllık bir Türk matematik öğretmenisin. Cevapların kısa, Türkçe ve örnekli olur."},

        {"role": "user", "content": "Üslü sayıyı 5. sınıf öğrencisine nasıl anlatırsın?"}

    ]

)

print(cevap["message"]["content"])

```

## Çoklu Turn (Memory)




```python
# Konuşma geçmişini elle yönet

history = [

    {"role": "system", "content": "Sen yardımsever bir asistanır. Türkçe konuş."},

]



def sor(soru):

    history.append({"role": "user", "content": soru})

    r = chat(model="llama3.2:3b", messages=history)

    cevap = r["message"]["content"]

    history.append({"role": "assistant", "content": cevap})

    return cevap



print(sor("Merhaba!"))

print(sor("Adın ne?"))

print(sor("Az önce ne sordum?"))

```

## Parametreler — temperature, top_p




```python
# Yaratıcılık vs tutarlılık

r = chat(

    model="llama3.2:3b",

    messages=[{"role": "user", "content": "Üretken YZ üzerine 4 satırlık bir şiir yaz."}],

    options={"temperature": 1.5, "top_p": 0.95}

)

print("Yaratıcı:", r["message"]["content"])



r = chat(

    model="llama3.2:3b",

    messages=[{"role": "user", "content": "Üretken YZ üzerine 4 satırlık bir şiir yaz."}],

    options={"temperature": 0.2, "top_p": 0.5}

)

print("Tutarlı:", r["message"]["content"])

```
