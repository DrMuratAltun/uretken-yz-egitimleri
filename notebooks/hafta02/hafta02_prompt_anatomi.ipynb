## 1. Yapı 1 — ROL

"Sen bir... uzmanısın." Modelin "kim olduğunu" bilmesi cevabın tonunu, derinliğini ve dilini belirler.

**Örnek:**
```
Sen 20 yıllık bir matematik öğretmenisin. 8. sınıf öğrencilerine üslü sayıları somut örneklerle anlatırsın.
```


## 2. Yapı 2 — BAĞLAM

Modele "neye bağlı" olduğunu söyle. Sınıf düzeyi, hedef kitle, kısıtlar, kurum kuralları.

**Örnek:**
```
Öğrenciler MEB müfredatına göre çalışıyor. Önümüzdeki hafta TEOG sınavı var. Konuyu hızlı pekiştirmem gerekiyor.
```


## 3. Yapı 3 — GÖREV

Net bir fiil + somut çıktı. "Yardım et" değil, "5 örnek soru hazırla".


## 4. Yapı 4 — KISIT

Sınırlar: dil, uzunluk, format, ton, yasak kelimeler.

**Örnek:**
```
- Türkçe (ı, ğ, ü, ö, ş, ç kullan)
- Her cevap maksimum 100 kelime
- Akademik dil değil, samimi öğretmen dili
```


## 5. Yapı 5 — FORMAT

JSON, tablo, markdown, listeleme. Yapılandırılmış çıktı kullanım kolaylığı sağlar.


## 6. Yapı 6 — ÖRNEK (Few-shot)

1-3 örnek vermek %80 işi çözer. "Örneğin böyle olsun: ..."
