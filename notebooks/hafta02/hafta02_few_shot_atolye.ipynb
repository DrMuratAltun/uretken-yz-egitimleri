## 1. Tanımlar

- **Zero-shot:** Hiç örnek vermeden görev. "Bu yorumun duygusunu söyle."
- **One-shot:** 1 örnek. "Şöyle: 'Harika!' = pozitif. Sıra sende: 'Berbat' = ?"
- **Few-shot:** 2-5 örnek. Modelin örüntüyü çıkarmasını sağlar.


## 2. Atölye 1 — Türkçe Yorum Duygu Analizi

ChatGPT veya Gemini'de aşağıdaki üç prompt'u dene ve sonuçları kaydet:

**Zero-shot:**
```
Bu Türkçe yorumun duygusu nedir (pozitif/negatif/nötr)?
Yorum: "Kargo geç geldi ama ürün güzel."
```

**Few-shot (3 örnek):**
```
Yorum: "Teslim çok hızlı, çok memnunum." → POZ
Yorum: "Renk fotoğraftaki gibi değil." → NEG
Yorum: "İdare eder, beklediğim kadar değil." → NÖT
Yorum: "Kargo geç geldi ama ürün güzel." →
```


## 3. Atölye 2 — Soru Sınıflandırma (TYT)

TYT soru tipi sınıflandırma:
```
Soru: "x² - 5x + 6 = 0 denklemini çözünüz." → CEBİR
Soru: "Türkiye'nin en büyük gölü hangisidir?" → COĞRAFYA
Soru: "Aşağıdakilerden hangisi atom çekirdeğinde bulunmaz?" →
```


## 4. İyi Few-shot Örnekleri Nasıl Seçilir?

- **Çeşitlilik:** Farklı tipte örnekler (kısa/uzun, pozitif/negatif/nötr)
- **Format tutarlılığı:** Aynı yapı (girdi → çıktı)
- **Sınır vakaları:** "İdare eder" gibi belirsizler dahil et
- **3-5 yeterli:** Çok örnek hız+maliyet getirir, az fayda
