## 2. Veri Hazırlığı: Bahşiş Oranı Hesaplama

Bahşiş miktarı tek başına karşılaştırma için yeterli değildir. Hesap tutarına göre **bahşiş yüzdesi** hesaplamamız gerekir.

```python
# Bahşiş yüzdesi sütunu oluştur
tips["tip_percentage"] = tips["tip"] / tips["total_bill"] * 100

print("Bahşiş yüzdesi (tip_percentage) eklendi:")
print(tips[["total_bill", "tip", "tip_percentage"]].head(10))
print(f"\nOrtalama bahşiş yüzdesi: %{tips['tip_percentage'].mean():.2f}")
print(f"Medyan bahşiş yüzdesi: %{tips['tip_percentage'].median():.2f}")
```

---
## 3. Soru 1: Sigara İçenler mi İçmeyenler mi Daha Çok Bahşiş Bırakıyor?

```python
# Sigara durumuna göre bahşiş istatistikleri
sigara_bahsis = tips.groupby("smoker").agg({
    "tip": ["mean", "median", "count"],
    "tip_percentage": ["mean", "median"]
}).round(2)

print("Sigara durumuna göre bahşiş analizi:")
print(sigara_bahsis)
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Ortalama bahşiş yüzdesi karşılaştırması
print("\nOrtalama bahşiş yüzdesi:")
print(tips.groupby("smoker")["tip_percentage"].mean().round(2))
print()

# Ortalama bahşiş miktarı
print("Ortalama bahşiş miktarı ($):")
print(tips.groupby("smoker")["tip"].mean().round(2))
```

### Yorum

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Yorum
icen_ort = tips[tips["smoker"] == "Yes"]["tip_percentage"].mean()
icmeyen_ort = tips[tips["smoker"] == "No"]["tip_percentage"].mean()

if icen_ort > icmeyen_ort:
    print(f"Sonuç: Sigara içenler ortalamada daha yüksek bahşiş yüzdesi bırakıyor (İçen: %{icen_ort:.2f}, İçmeyen: %{icmeyen_ort:.2f})")
else:
    print(f"Sonuç: Sigara içmeyenler ortalamada daha yüksek bahşiş yüzdesi bırakıyor (İçmeyen: %{icmeyen_ort:.2f}, İçen: %{icen_ort:.2f})")
```

---
## 4. Soru 2: Hafta Sonu mu Hafta İçi mi Daha Yoğun?

```python
# Hafta sonu/hafta içi sütunu oluştur
tips["hafta_sonu"] = tips["day"].apply(
    lambda x: "Hafta Sonu" if x in ["Sat", "Sun"] else "Hafta İçi"
)

print("Hafta sonu vs. hafta içi müşteri sayısı:")
print(tips["hafta_sonu"].value_counts())
print()

print("Yüzdelik dağılım:")
print(tips["hafta_sonu"].value_counts(normalize=True).map(lambda x: f"%{x*100:.1f}"))
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Hafta sonu vs. hafta içi karşılaştırması
print("Hafta sonu/içi ortalama hesap ve bahşiş:")
print(tips.groupby("hafta_sonu").agg({
    "total_bill": "mean",
    "tip": "mean",
    "tip_percentage": "mean",
    "size": ["mean", "count"]
}).round(2))
```

---
## 5. Soru 3: Kadınlar mı Erkekler mi Daha Yüksek Bahşiş Oranı Veriyor?

```python
# Cinsiyete göre bahşiş analizi
cinsiyet_bahsis = tips.groupby("sex").agg({
    "tip": ["mean", "median", "sum", "count"],
    "tip_percentage": ["mean", "median"],
    "total_bill": "mean"
}).round(2)

print("Cinsiyete göre bahşiş analizi:")
print(cinsiyet_bahsis)
```

### Detaylı karşılaştırma

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Detaylı karşılaştırma
erkek_ort = tips[tips["sex"] == "Male"]["tip_percentage"].mean()
kadin_ort = tips[tips["sex"] == "Female"]["tip_percentage"].mean()

print(f"Erkeklerin ortalama bahşiş yüzdesi: %{erkek_ort:.2f}")
print(f"Kadınların ortalama bahşiş yüzdesi: %{kadin_ort:.2f}")
print()

if kadin_ort > erkek_ort:
    print(f"Sonuç: Kadınlar %{kadin_ort - erkek_ort:.2f} puan daha yüksek bahşiş oranı veriyor.")
else:
    print(f"Sonuç: Erkekler %{erkek_ort - kadin_ort:.2f} puan daha yüksek bahşiş oranı veriyor.")
```

---
## 6. Soru 4: En Yoğun Gün Hangisi?

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Günlere göre müşteri sayısı
gun_dagilim = tips["day"].value_counts()
print("Günlere göre müşteri sayısı:")
print(gun_dagilim)
print(f"\nEn yoğun gün: {gun_dagilim.index[0]} ({gun_dagilim.values[0]} müşteri)")
print(f"En sakin gün: {gun_dagilim.index[-1]} ({gun_dagilim.values[-1]} müşteri)")
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Günlere göre detaylı istatistikler
gun_istatistik = tips.groupby("day").agg({
    "total_bill": ["mean", "sum"],
    "tip": ["mean", "sum"],
    "tip_percentage": "mean",
    "size": ["mean", "count"]
}).round(2)

print("Günlere göre detaylı istatistikler:")
print(gun_istatistik)
```

---
## 7. Soru 5: Bahşiş Oranı ile Hesap Tutarı Arasında İlişki Var mı?

```python
# Korelasyon analizi
korelasyon = tips[["total_bill", "tip", "tip_percentage", "size"]].corr().round(3)
print("Korelasyon matrisi:")
print(korelasyon)
```

### Hesap tutarı ile bahşiş arasındaki korelasyon

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Hesap tutarı ile bahşiş arasındaki korelasyon
kor_tutar_bahsis = tips["total_bill"].corr(tips["tip"])
kor_tutar_yuzde = tips["total_bill"].corr(tips["tip_percentage"])

print(f"Hesap tutarı ↔ Bahşiş miktarı korelasyonu: {kor_tutar_bahsis:.3f}")
print(f"Hesap tutarı ↔ Bahşiş yüzdesi korelasyonu: {kor_tutar_yuzde:.3f}")
print()

if kor_tutar_bahsis > 0.5:
    print("Yorum: Hesap tutarı arttıkça bahşiş miktarı da artıyor (güçlü pozitif ilişki).")

if kor_tutar_yuzde < 0:
    print("Yorum: Ancak hesap tutarı arttıkça bahşiş YÜZDESI düşüyor. Yani büyük hesaplarda bahşiş miktarı artsa da oranı azalıyor.")
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Hesap tutarına göre gruplar oluşturup bahşiş yüzdesini inceleyelim
tips["hesap_grubu"] = pd.cut(
    tips["total_bill"],
    bins=[0, 15, 25, 35, 60],
    labels=["Düşük (0-15$)", "Orta (15-25$)", "Yüksek (25-35$)", "Çok Yüksek (35$+)"]
)

print("Hesap grubuna göre bahşiş analizi:")
hesap_analiz = tips.groupby("hesap_grubu", observed=True).agg({
    "tip": "mean",
    "tip_percentage": "mean",
    "total_bill": "count"
}).round(2)

hesap_analiz.columns = ["Ort. Bahşiş ($)", "Ort. Bahşiş Yüzdesi (%)", "Müşteri Sayısı"]
print(hesap_analiz)
```

## 8. Ek Analizler

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Öğün zamanına göre analiz (Öğle vs. Akşam)
print("Öğün zamanına göre analiz:")
ogun_analiz = tips.groupby("time").agg({
    "total_bill": ["mean", "count"],
    "tip": "mean",
    "tip_percentage": "mean"
}).round(2)

print(ogun_analiz)
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Masa büyüklüğüne göre analiz
print("Masa büyüklüğüne (kişi sayısı) göre analiz:")
masa_analiz = tips.groupby("size").agg({
    "total_bill": "mean",
    "tip": "mean",
    "tip_percentage": "mean",
    "size": "count"
})

masa_analiz.columns = ["Ort. Hesap ($)", "Ort. Bahşiş ($)", "Ort. Bahşiş Yüzdesi (%)", "Masa Sayısı"]
print(masa_analiz.round(2))
```

### En cömert ve en cimri müşteriler

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# En cömert ve en cimri müşteriler
print("En yüksek bahşiş yüzdesi veren 5 müşteri:")
print(tips.nlargest(5, "tip_percentage")[["total_bill", "tip", "tip_percentage", "sex", "day"]])
print()

print("En düşük bahşiş yüzdesi veren 5 müşteri:")
print(tips.nsmallest(5, "tip_percentage")[["total_bill", "tip", "tip_percentage", "sex", "day"]])
```
