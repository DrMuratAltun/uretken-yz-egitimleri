## 1. Şablon Yapısı

Her prompt için minimum alanlar:

- **Başlık** (1 satır)
- **Etiket** (#egitimci, #pazarlama, #kod...)
- **Kullanım Sıklığı** (★★★)
- **Versiyon** (v1.0)
- **Prompt metni**
- **Örnek Çıktı**
- **Değişkenler** ({{konu}}, {{seviye}})
- **Notlar** (kim için, hangi araçta test edildi)


## 2. Notion Şablonu

Notion'da Database aç → her prompt bir kart. Filtre: etiket. Sırala: kullanım sıklığı. Paylaş: takım için "Public" link.


## 3. GitHub Şablonu

Bir repo aç (`prompt-kutuphanesi`). Klasör yapısı:
```
/prompts/
  /egitimci/
    /ders-plani.md
    /soru-uretici.md
  /kod/
  /yazilim/
```


## 4. Custom GPT / Gem'e Aktarma

En sık kullandığın 5 prompt'u Custom GPT veya Gem'e dönüştür. Tek tıkla aktif et.
