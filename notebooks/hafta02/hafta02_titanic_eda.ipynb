## 3. Eksik Veri Analizi

### Veri Ön İşleme

Modeli eğitmeden önce veriyi temizliyoruz ve dönüştürüyoruz. Eksik değerleri doldurma, kategorik değişkenleri sayıya çevirme ve ölçeklendirme bu adımın parçasıdır.

```python
# Eksik değer sayısı
eksik = titanic.isnull().sum()
eksik_yuzde = (titanic.isnull().sum() / len(titanic) * 100).round(2)

eksik_tablo = pd.DataFrame({
    "Eksik Sayısı": eksik,
    "Eksik Yüzdesi (%)": eksik_yuzde
})

# Sadece eksik olanları göster
eksik_tablo = eksik_tablo[eksik_tablo["Eksik Sayısı"] > 0].sort_values("Eksik Sayısı", ascending=False)
print("Eksik değer içeren sütunlar:")
print(eksik_tablo)
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Eksik veri görselleştirmesi
fig, ax = plt.subplots(figsize=(10, 4))
eksik_tablo["Eksik Yüzdesi (%)"].plot(kind="barh", color="coral", ax=ax)
ax.set_title("Eksik Veri Oranları", fontsize=14, fontweight="bold")
ax.set_xlabel("Eksik Yüzde (%)")
plt.tight_layout()
plt.show()
```

## 4. Genel Hayatta Kalma Dağılımı

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Hayatta kalma sayıları
print("Hayatta kalma dağılımı:")
print(titanic["survived"].value_counts())
print()
print("Yüzdelik:")
print(titanic["survived"].value_counts(normalize=True).map(lambda x: f"%{x*100:.1f}"))
print(f"\nToplam yolcu: {len(titanic)}")
print(f"Hayatta kalanlar: {titanic['survived'].sum()} (%{titanic['survived'].mean()*100:.1f})")
print(f"Hayatını kaybedenler: {(titanic['survived'] == 0).sum()} (%{(1-titanic['survived'].mean())*100:.1f})")
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Hayatta kalma grafiği
fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=titanic, x="survived", palette=["#e74c3c", "#2ecc71"], ax=ax)
ax.set_title("Hayatta Kalma Dağılımı", fontsize=14, fontweight="bold")
ax.set_xlabel("Hayatta Kaldı mı? (0=Hayır, 1=Evet)")
ax.set_ylabel("Yolcu Sayısı")
ax.set_xticklabels(["Hayatını Kaybetti", "Hayatta Kaldı"])

# Çubukların üstüne sayı ekle
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```

## 5. Soru 1: Kadınlar mı Erkekler mi Daha Çok Hayatta Kaldı?

```python
# Cinsiyete göre hayatta kalma oranları
cinsiyet_hayatta = titanic.groupby("sex")["survived"].agg(["sum", "count", "mean"]).round(3)
cinsiyet_hayatta.columns = ["Hayatta Kalan", "Toplam", "Hayatta Kalma Oranı"]
print("Cinsiyete göre hayatta kalma:")
print(cinsiyet_hayatta)
print()

erkek_oran = titanic[titanic["sex"] == "male"]["survived"].mean()
kadin_oran = titanic[titanic["sex"] == "female"]["survived"].mean()
print(f"Kadınların hayatta kalma oranı: %{kadin_oran*100:.1f}")
print(f"Erkeklerin hayatta kalma oranı: %{erkek_oran*100:.1f}")
print(f"\nKadınların hayatta kalma şansı erkeklerin {kadin_oran/erkek_oran:.1f} katı!")
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Cinsiyet bazlı hayatta kalma grafiği
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Sol: Sayısal dağılım
sns.countplot(data=titanic, x="sex", hue="survived", palette=["#e74c3c", "#2ecc71"], ax=axes[0])
axes[0].set_title("Cinsiyete Göre Hayatta Kalma (Sayı)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Cinsiyet")
axes[0].set_ylabel("Yolcu Sayısı")
axes[0].legend(["Hayatını Kaybetti", "Hayatta Kaldı"])

# Sağ: Oran
cinsiyet_oran = titanic.groupby("sex")["survived"].mean()
cinsiyet_oran.plot(kind="bar", color=["#3498db", "#e91e63"], ax=axes[1])
axes[1].set_title("Cinsiyete Göre Hayatta Kalma Oranı", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Cinsiyet")
axes[1].set_ylabel("Hayatta Kalma Oranı")
axes[1].set_ylim(0, 1)
axes[1].set_xticklabels(["Erkek", "Kadın"], rotation=0)

for i, v in enumerate(cinsiyet_oran.values):
    axes[1].text(i, v + 0.02, f"%{v*100:.1f}", ha='center', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```

## 6. Soru 2: Hangi Sınıf Bilet Alanlar Daha Çok Kurtuldu?

```python
# Sınıfa göre hayatta kalma oranları
sinif_hayatta = titanic.groupby("pclass")["survived"].agg(["sum", "count", "mean"]).round(3)
sinif_hayatta.columns = ["Hayatta Kalan", "Toplam", "Hayatta Kalma Oranı"]
sinif_hayatta.index = ["1. Sınıf", "2. Sınıf", "3. Sınıf"]
print("Bilet sınıfına göre hayatta kalma:")
print(sinif_hayatta)
```

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Sınıf bazlı hayatta kalma grafiği
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Sol: Sayısal
sns.countplot(data=titanic, x="pclass", hue="survived", palette=["#e74c3c", "#2ecc71"], ax=axes[0])
axes[0].set_title("Bilet Sınıfına Göre Hayatta Kalma (Sayı)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Bilet Sınıfı")
axes[0].set_ylabel("Yolcu Sayısı")
axes[0].legend(["Hayatını Kaybetti", "Hayatta Kaldı"])

# Sağ: Oran
sinif_oran = titanic.groupby("pclass")["survived"].mean()
sinif_oran.plot(kind="bar", color=["#f1c40f", "#3498db", "#95a5a6"], ax=axes[1])
axes[1].set_title("Bilet Sınıfına Göre Hayatta Kalma Oranı", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Bilet Sınıfı")
axes[1].set_ylabel("Hayatta Kalma Oranı")
axes[1].set_ylim(0, 1)
axes[1].set_xticklabels(["1. Sınıf", "2. Sınıf", "3. Sınıf"], rotation=0)

for i, v in enumerate(sinif_oran.values):
    axes[1].text(i, v + 0.02, f"%{v*100:.1f}", ha='center', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```

### Veri Gruplama ve Analiz

```python
# Sınıf + cinsiyet birlikte
sinif_cinsiyet = titanic.groupby(["pclass", "sex"])["survived"].mean().round(3).unstack()
print("Sınıf ve cinsiyete göre hayatta kalma oranı:")
print(sinif_cinsiyet)
```

## 7. Soru 3: Yaş Dağılımı Nasıl?

### Temel İstatistikler

Verinin genel yapısını inceliyoruz: sütun tipleri, eksik değerler, temel istatistikler (ortalama, medyan, min, max). Bu bilgiler veri temizleme ve ön işleme adımlarını planlamak için gereklidir.

```python
# Yaş istatistikleri
print("Yaş istatistikleri:")
print(titanic["age"].describe().round(2))
print(f"\nEksik yaş sayısı: {titanic['age'].isnull().sum()} (%{titanic['age'].isnull().mean()*100:.1f})")
```

### Dağılım Görselleştirmesi

Aşağıdaki grafikte verinin dağılımını histogram ile inceliyoruz. Dağılımın şekli (normal, çarpık, bimodal) hangi istatistiksel yöntemlerin uygulanabileceğini belirler.

```python
# Yaş dağılımı histogramı
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sol: Genel yaş dağılımı
titanic["age"].hist(bins=30, color="#3498db", edgecolor="white", ax=axes[0])
axes[0].set_title("Yaş Dağılımı", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Yaş")
axes[0].set_ylabel("Yolcu Sayısı")
axes[0].axvline(titanic["age"].mean(), color="red", linestyle="--", label=f'Ortalama: {titanic["age"].mean():.1f}')
axes[0].axvline(titanic["age"].median(), color="green", linestyle="--", label=f'Medyan: {titanic["age"].median():.1f}')
axes[0].legend()

# Sağ: Hayatta kalma durumuna göre yaş
titanic[titanic["survived"] == 0]["age"].hist(bins=30, alpha=0.6, color="#e74c3c", label="Hayatını Kaybetti", ax=axes[1])
titanic[titanic["survived"] == 1]["age"].hist(bins=30, alpha=0.6, color="#2ecc71", label="Hayatta Kaldı", ax=axes[1])
axes[1].set_title("Hayatta Kalma Durumuna Göre Yaş", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Yaş")
axes[1].set_ylabel("Yolcu Sayısı")
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Yaş grupları oluşturma
titanic["yas_grubu"] = pd.cut(
    titanic["age"],
    bins=[0, 12, 18, 35, 55, 80],
    labels=["Çocuk (0-12)", "Genç (13-18)", "Yetişkin (19-35)", "Orta Yaş (36-55)", "Yaşlı (56+)"]
)

yas_analiz = titanic.groupby("yas_grubu", observed=True)["survived"].agg(["sum", "count", "mean"]).round(3)
yas_analiz.columns = ["Hayatta Kalan", "Toplam", "Hayatta Kalma Oranı"]
print("Yaş grubuna göre hayatta kalma:")
print(yas_analiz)
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Yaş grubuna göre hayatta kalma grafiği
fig, ax = plt.subplots(figsize=(10, 5))
yas_oran = titanic.groupby("yas_grubu", observed=True)["survived"].mean()
yas_oran.plot(kind="bar", color=["#e74c3c", "#e67e22", "#3498db", "#9b59b6", "#1abc9c"], ax=ax)
ax.set_title("Yaş Grubuna Göre Hayatta Kalma Oranı", fontsize=14, fontweight="bold")
ax.set_xlabel("Yaş Grubu")
ax.set_ylabel("Hayatta Kalma Oranı")
ax.set_ylim(0, 1)
plt.xticks(rotation=30)

for i, v in enumerate(yas_oran.values):
    ax.text(i, v + 0.02, f"%{v*100:.1f}", ha='center', fontweight='bold', fontsize=11)

plt.tight_layout()
plt.show()
```

## 8. Soru 4: Aile Büyüklüğü Hayatta Kalmayı Etkiliyor mu?

```python
# Aile büyüklüğü grafiği
fig, ax = plt.subplots(figsize=(10, 5))
aile_oran = titanic.groupby("aile_buyuklugu")["survived"].mean()
aile_oran.plot(kind="bar", color="#3498db", ax=ax)
ax.set_title("Aile Büyüklüğüne Göre Hayatta Kalma Oranı", fontsize=14, fontweight="bold")
ax.set_xlabel("Aile Büyüklüğü (kişi)")
ax.set_ylabel("Hayatta Kalma Oranı")
ax.set_ylim(0, 1)
plt.xticks(rotation=0)

for i, v in enumerate(aile_oran.values):
    ax.text(i, v + 0.02, f"%{v*100:.0f}", ha='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()
```

### Veri Gruplama ve Analiz

## 10. Korelasyon Analizi

### Sayısal sütunların korelasyonu

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Sayısal sütunların korelasyonu
sayisal_sutunlar = ["survived", "pclass", "age", "sibsp", "parch", "fare"]
korelasyon = titanic[sayisal_sutunlar].corr().round(3)

print("Korelasyon matrisi:")
print(korelasyon)
```

### Korelasyon Analizi ve Isı Haritası

Özellikler arasındaki ilişkiyi korelasyon matrisi ve ısı haritası ile görselleştiriyoruz. Yüksek korelasyonlu özellikler modelin en çok yararlanacağı özelliklerdir.

```python
# Korelasyon ısı haritası
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(korelasyon, annot=True, cmap="RdBu_r", center=0, fmt=".2f",
            linewidths=0.5, ax=ax)
ax.set_title("Korelasyon Isı Haritası", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Hayatta kalma ile en yüksek korelasyona sahip değişkenler
hayatta_kor = korelasyon["survived"].drop("survived").abs().sort_values(ascending=False)
print("Hayatta kalma ile en güçlü korelasyonlar (mutlak değer):")
for col, val in hayatta_kor.items():
    yon = "pozitif" if korelasyon.loc["survived", col] > 0 else "negatif"
    print(f"  {col}: {val:.3f} ({yon})")
```

## 11. Özet ve Bulgular

Bu analizde **891 Titanic yolcusunun** verilerini inceledik.

### Temel Bulgular:

1. **Cinsiyet**: Kadınların hayatta kalma oranı erkeklerden çok daha yüksek. "Önce kadınlar ve çocuklar" kuralı verilere yansıyor.

2. **Bilet sınıfı**: 1. sınıf yolcuların hayatta kalma oranı en yüksek, 3. sınıfın en düşük. Sosyo-ekonomik statü hayatta kalmada belirleyici.

3. **Yaş**: Çocuklar (0-12) en yüksek hayatta kalma oranına sahip. Yaş arttıkça oran düşüyor.

4. **Aile büyüklüğü**: 2-4 kişilik ailelerin hayatta kalma oranı yalnız yolculardan ve çok büyük ailelerden daha yüksek.

### Kullanılan Teknikler:
- `groupby()` ile gruplama ve karşılaştırma
- `value_counts()` ile dağılım analizi
