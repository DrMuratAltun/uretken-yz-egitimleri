Projede kullanacağımız kütüphaneleri içe aktarıyoruz:

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from urllib.request import urlretrieve
import os

print(f"OpenCV sürümü: {cv2.__version__}")
print(f"NumPy sürümü: {np.__version__}")
```

## 2. NumPy ile Görüntü Oluşturma

Görüntüler aslında çok boyutlu NumPy dizileridir:
- **Gri tonlama:** (yükseklik, genişlik) - 2 boyutlu
- **Renkli (BGR/RGB):** (yükseklik, genişlik, 3) - 3 boyutlu
- Piksel değerleri: 0-255 arası (uint8)

```python
# Farklı renkli dikdörtgenler oluştur
img_renkli = np.zeros((300, 400, 3), dtype=np.uint8)

# Kırmızı dikdörtgen (RGB formatında)
img_renkli[20:100, 20:180] = [255, 0, 0]

# Yeşil dikdörtgen
img_renkli[50:150, 120:300] = [0, 255, 0]

# Mavi dikdörtgen
img_renkli[120:250, 200:380] = [0, 0, 255]

# Sarı dikdörtgen
img_renkli[180:280, 30:200] = [255, 255, 0]

plt.figure(figsize=(8, 6))
plt.imshow(img_renkli)
plt.title('NumPy ile Oluşturulan Renkli Dikdörtgenler', fontsize=13, fontweight='bold')
plt.axis('off')
plt.show()

print(f"Görüntü boyutu: {img_renkli.shape}")
print(f"Veri tipi: {img_renkli.dtype}")
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Renk geçişi (gradient) oluştur
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Yatay gradient
gradient_yatay = np.zeros((200, 300), dtype=np.uint8)
for j in range(300):
    gradient_yatay[:, j] = int(j / 300 * 255)

axes[0].imshow(gradient_yatay, cmap='gray')
axes[0].set_title('Yatay Gradient', fontsize=12, fontweight='bold')
axes[0].axis('off')

# Dikey gradient
gradient_dikey = np.zeros((200, 300), dtype=np.uint8)
for i in range(200):
    gradient_dikey[i, :] = int(i / 200 * 255)

axes[1].imshow(gradient_dikey, cmap='gray')
axes[1].set_title('Dikey Gradient', fontsize=12, fontweight='bold')
axes[1].axis('off')

# Renkli gradient (RGB)
gradient_rgb = np.zeros((200, 300, 3), dtype=np.uint8)
for j in range(300):
    gradient_rgb[:, j, 0] = int(j / 300 * 255)         # Kırmızı
    gradient_rgb[:, j, 1] = int((300 - j) / 300 * 255)  # Yeşil
    gradient_rgb[:, j, 2] = 128                          # Mavi sabit

axes[2].imshow(gradient_rgb)
axes[2].set_title('Renkli Gradient', fontsize=12, fontweight='bold')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## 3. Görüntü Okuma ve Görüntüleme

**Dikkat:** OpenCV görüntüleri **BGR** formatında okur, matplotlib ise **RGB** bekler. Bu yüzden dönüşüm gerekir.

```python
# Örnek bir görüntü indir (Colab uyumlu)
url = "https://upload.wikimedia.org/wikipedia/commons/thumb/4/47/PNG_transparency_demonstration_1.png/280px-PNG_transparency_demonstration_1.png"
dosya_adi = "ornek_goruntu.png"

if not os.path.exists(dosya_adi):
    urlretrieve(url, dosya_adi)
    print(f"Görüntü indirildi: {dosya_adi}")

# OpenCV ile oku (BGR formatı)
img_bgr = cv2.imread(dosya_adi)

if img_bgr is not None:
    print(f"Görüntü boyutu: {img_bgr.shape}")
    print(f"Veri tipi: {img_bgr.dtype}")
    print(f"Toplam piksel sayısı: {img_bgr.size}")
    
    # BGR → RGB dönüşümü
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    ax1.imshow(img_bgr)  # BGR - renkler yanlış görünür
    ax1.set_title('BGR Formatı (Yanlış Renkler)', fontsize=12, fontweight='bold')
    ax1.axis('off')
    
    ax2.imshow(img_rgb)  # RGB - doğru renkler
    ax2.set_title('RGB Formatı (Doğru Renkler)', fontsize=12, fontweight='bold')
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("Görüntü yüklenemedi. Sentetik görüntü kullanılıyor.")
    # Sentetik renkli görüntü oluştur
    img_bgr = np.random.randint(0, 256, (300, 400, 3), dtype=np.uint8)
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(6, 4))
    plt.imshow(img_rgb)
    plt.title('Sentetik Görüntü')
    plt.axis('off')
    plt.show()
```

## 4. Renk Uzayları

OpenCV birçok renk uzayını destekler:
- **BGR:** Varsayılan (Mavi, Yeşil, Kırmızı)
- **RGB:** Standart (Kırmızı, Yeşil, Mavi)
- **GRAY:** Gri tonlama (tek kanal)
- **HSV:** Ton, Doygunluk, Değer (renk filtreleme için ideal)

```python
# Test görüntüsü oluştur (renkli şekiller)
test_img = np.zeros((300, 400, 3), dtype=np.uint8)
cv2.circle(test_img, (100, 150), 80, (255, 0, 0), -1)   # Mavi daire (BGR)
cv2.circle(test_img, (200, 150), 80, (0, 255, 0), -1)    # Yeşil daire
cv2.circle(test_img, (300, 150), 80, (0, 0, 255), -1)    # Kırmızı daire

# Renk uzayı dönüşümleri
img_rgb_test = cv2.cvtColor(test_img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(test_img, cv2.COLOR_BGR2GRAY)
img_hsv = cv2.cvtColor(test_img, cv2.COLOR_BGR2HSV)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].imshow(img_rgb_test)
axes[0, 0].set_title(f'RGB - Boyut: {img_rgb_test.shape}', fontsize=12, fontweight='bold')
axes[0, 0].axis('off')

axes[0, 1].imshow(img_gray, cmap='gray')
axes[0, 1].set_title(f'Gri Tonlama - Boyut: {img_gray.shape}', fontsize=12, fontweight='bold')
axes[0, 1].axis('off')

# HSV kanallarını ayrı göster
axes[1, 0].imshow(img_hsv[:, :, 0], cmap='hsv')
axes[1, 0].set_title('HSV - Ton (Hue) Kanalı', fontsize=12, fontweight='bold')
axes[1, 0].axis('off')

axes[1, 1].imshow(img_hsv[:, :, 1], cmap='gray')
axes[1, 1].set_title('HSV - Doygunluk (Saturation) Kanalı', fontsize=12, fontweight='bold')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

## 6. Piksel Erişimi ve Değiştirme

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Örnek görüntü oluştur
img_piksel = np.zeros((200, 300, 3), dtype=np.uint8)
img_piksel[:, :] = [100, 150, 200]  # Açık mavi arka plan

# Tek piksel oku
piksel = img_piksel[100, 150]
print(f"(100, 150) konumundaki piksel değeri (BGR): {piksel}")

# Belirli bir bölgeyi değiştir
img_piksel[50:80, 50:100] = [0, 0, 255]    # Kırmızı kare
img_piksel[50:80, 120:170] = [0, 255, 0]    # Yeşil kare
img_piksel[50:80, 190:240] = [255, 0, 0]    # Mavi kare

# Piksel piksel desen oluştur (satranç tahtası)
for i in range(100, 180, 10):
    for j in range(50, 250, 10):
        if (i // 10 + j // 10) % 2 == 0:
            img_piksel[i:i+10, j:j+10] = [255, 255, 255]
        else:
            img_piksel[i:i+10, j:j+10] = [0, 0, 0]

plt.figure(figsize=(8, 5))
plt.imshow(cv2.cvtColor(img_piksel, cv2.COLOR_BGR2RGB))
plt.title('Piksel Manipülasyonu Örneği', fontsize=13, fontweight='bold')
plt.axis('off')
plt.show()
```

## 7. Görüntü Kırpma ve Yeniden Boyutlandırma

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Renkli bir test görüntüsü oluştur
img_orjinal = np.zeros((400, 600, 3), dtype=np.uint8)
# Renkli bölgeler
img_orjinal[0:200, 0:300] = [255, 100, 50]    # Sol üst
img_orjinal[0:200, 300:600] = [50, 255, 100]   # Sağ üst
img_orjinal[200:400, 0:300] = [100, 50, 255]   # Sol alt
img_orjinal[200:400, 300:600] = [200, 200, 50]  # Sağ alt
cv2.circle(img_orjinal, (300, 200), 100, (255, 255, 255), 3)

# Kırpma (ROI - Region of Interest)
kirpilmis = img_orjinal[50:250, 150:450]  # [y1:y2, x1:x2]

# Yeniden boyutlandırma
boyut_kucuk = cv2.resize(img_orjinal, (150, 100))  # (genişlik, yükseklik)
boyut_buyuk = cv2.resize(kirpilmis, (600, 400), interpolation=cv2.INTER_CUBIC)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].imshow(cv2.cvtColor(img_orjinal, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title(f'Orijinal: {img_orjinal.shape[:2]}', fontsize=12, fontweight='bold')
axes[0, 0].axis('off')

axes[0, 1].imshow(cv2.cvtColor(kirpilmis, cv2.COLOR_BGR2RGB))
axes[0, 1].set_title(f'Kırpılmış (ROI): {kirpilmis.shape[:2]}', fontsize=12, fontweight='bold')
axes[0, 1].axis('off')

axes[1, 0].imshow(cv2.cvtColor(boyut_kucuk, cv2.COLOR_BGR2RGB))
axes[1, 0].set_title(f'Küçültülmüş: {boyut_kucuk.shape[:2]}', fontsize=12, fontweight='bold')
axes[1, 0].axis('off')

axes[1, 1].imshow(cv2.cvtColor(boyut_buyuk, cv2.COLOR_BGR2RGB))
axes[1, 1].set_title(f'Büyütülmüş (Cubic): {boyut_buyuk.shape[:2]}', fontsize=12, fontweight='bold')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()

print("\nYeniden boyutlandırma yöntemleri:")
print("  cv2.INTER_NEAREST - En yakın komşu (hızlı, kalitesiz)")
print("  cv2.INTER_LINEAR  - Bilineer (varsayılan, dengeli)")
print("  cv2.INTER_CUBIC   - Bikübik (yavaş, kaliteli)")
print("  cv2.INTER_AREA    - Alan tabanlı (küçültme için ideal)")
```

## 8. Kanal Ayırma ve Birleştirme

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Renkli gradient görüntüsü oluştur
img_kanal = np.zeros((300, 400, 3), dtype=np.uint8)
for i in range(300):
    for j in range(400):
        img_kanal[i, j] = [
            int(j / 400 * 255),       # Mavi: soldan sağa artar
            int(i / 300 * 255),        # Yeşil: yukarıdan aşağı artar
            int((400 - j) / 400 * 255)  # Kırmızı: sağdan sola artar
        ]

# Kanalları ayır
b, g, r = cv2.split(img_kanal)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].imshow(cv2.cvtColor(img_kanal, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title('Orijinal (RGB)', fontsize=12, fontweight='bold')
axes[0, 0].axis('off')

axes[0, 1].imshow(r, cmap='Reds')
axes[0, 1].set_title('Kırmızı Kanalı (R)', fontsize=12, fontweight='bold')
axes[0, 1].axis('off')

axes[1, 0].imshow(g, cmap='Greens')
axes[1, 0].set_title('Yeşil Kanalı (G)', fontsize=12, fontweight='bold')
axes[1, 0].axis('off')

axes[1, 1].imshow(b, cmap='Blues')
axes[1, 1].set_title('Mavi Kanalı (B)', fontsize=12, fontweight='bold')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Kanalları farklı sırada birleştir
img_birlesik = cv2.merge([r, g, b])  # RGB sırasında birleştir
img_ters = cv2.merge([g, b, r])      # Farklı sıra

# Tek kanal vurgulama
sadece_kirmizi = np.zeros_like(img_kanal)
sadece_kirmizi[:, :, 2] = r  # Sadece kırmızı kanalı koru

sadece_yesil = np.zeros_like(img_kanal)
sadece_yesil[:, :, 1] = g  # Sadece yeşil kanalı koru

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].imshow(cv2.cvtColor(sadece_kirmizi, cv2.COLOR_BGR2RGB))
axes[0].set_title('Sadece Kırmızı Kanal', fontsize=12, fontweight='bold')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(sadece_yesil, cv2.COLOR_BGR2RGB))
axes[1].set_title('Sadece Yeşil Kanal', fontsize=12, fontweight='bold')
axes[1].axis('off')

axes[2].imshow(cv2.cvtColor(img_ters, cv2.COLOR_BGR2RGB))
axes[2].set_title('Karıştırılmış Kanallar', fontsize=12, fontweight='bold')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
