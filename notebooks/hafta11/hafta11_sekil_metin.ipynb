### Kütüphanelerin Yüklenmesi

Projede kullanacağımız kütüphaneleri içe aktarıyoruz:

| Kütüphane | Amacı |
|-----------|-------|
| `cv2` | Bilgisayarlı görü ve görüntü işleme (OpenCV) |
| `matplotlib` | Grafik ve görselleştirme |
| `numpy` | Sayısal hesaplamalar ve dizi işlemleri |


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def goster(basliklar, goruntular, cmap_listesi=None, satir=1):
    """Birden fazla görüntüyü yan yana gösterir."""
    n = len(goruntular)
    sutun = (n + satir - 1) // satir
    fig, axes = plt.subplots(satir, sutun, figsize=(5 * sutun, 5 * satir))
    if satir == 1 and sutun == 1:
        axes = np.array([axes])
    axes = np.array(axes).flatten()
    for i in range(n):
        if len(goruntular[i].shape) == 3:
            axes[i].imshow(cv2.cvtColor(goruntular[i], cv2.COLOR_BGR2RGB))
        else:
            cmap = cmap_listesi[i] if cmap_listesi else 'gray'
            axes[i].imshow(goruntular[i], cmap=cmap)
        axes[i].set_title(basliklar[i], fontsize=11, fontweight='bold')
        axes[i].axis('off')
    for j in range(n, len(axes)):
        axes[j].axis('off')
    plt.tight_layout()
    plt.show()

print(f"OpenCV sürümü: {cv2.__version__}")
```

## 2. Boş Tuval Oluşturma

### Siyah tuval

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

## 3. Temel Şekil Çizimi

### 3.1 Dikdörtgen: `cv2.rectangle()`

```python
img = np.zeros((500, 500, 3), dtype=np.uint8)
img[:] = [40, 40, 40]

# Dolgulu dikdörtgen (kalınlık=-1)
cv2.rectangle(img, (30, 30), (200, 150), (0, 0, 255), -1)  # Kırmızı (BGR)

# Çerçeve dikdörtgen
cv2.rectangle(img, (250, 30), (470, 150), (0, 255, 0), 3)  # Yeşil, 3px

# Farklı kalınlıklarda
cv2.rectangle(img, (30, 200), (150, 320), (255, 0, 0), 1)   # Mavi, 1px
cv2.rectangle(img, (180, 200), (300, 320), (255, 0, 0), 5)   # Mavi, 5px
cv2.rectangle(img, (330, 200), (470, 320), (255, 0, 0), 10)  # Mavi, 10px

# Yarı saydam efekt
overlay = img.copy()
cv2.rectangle(overlay, (100, 350), (400, 480), (0, 255, 255), -1)
img = cv2.addWeighted(overlay, 0.5, img, 0.5, 0)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('cv2.rectangle() Örnekleri', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

### 3.2 Daire: `cv2.circle()`

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
img = np.zeros((500, 500, 3), dtype=np.uint8)
img[:] = [40, 40, 40]

# Dolgulu daire
cv2.circle(img, (150, 150), 80, (0, 165, 255), -1)  # Turuncu

# Çerçeve daireler
cv2.circle(img, (350, 150), 80, (255, 255, 0), 2)    # Cyan
cv2.circle(img, (350, 150), 60, (255, 255, 0), 2)
cv2.circle(img, (350, 150), 40, (255, 255, 0), 2)
cv2.circle(img, (350, 150), 20, (255, 255, 0), 2)

# İç içe renkli daireler
renkler = [(0, 0, 200), (0, 200, 0), (200, 0, 0), (0, 200, 200), (200, 0, 200)]
for i, (r, renk) in enumerate(zip([100, 80, 60, 40, 20], renkler)):
    cv2.circle(img, (250, 380), r, renk, -1)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('cv2.circle() Örnekleri', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

### 3.3 Çizgi ve Elips

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
img = np.zeros((500, 500, 3), dtype=np.uint8)
img[:] = [40, 40, 40]

# Çizgiler
cv2.line(img, (50, 50), (450, 50), (0, 255, 0), 2)      # Yatay
cv2.line(img, (50, 50), (50, 450), (0, 0, 255), 2)       # Dikey
cv2.line(img, (50, 100), (450, 250), (255, 255, 0), 3)   # Çapraz

# Ok çizgisi
cv2.arrowedLine(img, (100, 300), (400, 300), (255, 0, 255), 3, tipLength=0.05)

# Elips
cv2.ellipse(img, (250, 400), (150, 60), 0, 0, 360, (0, 255, 255), 2)    # Tam elips
cv2.ellipse(img, (250, 400), (100, 40), 30, 0, 360, (255, 128, 0), 2)   # Döndürülmüş
cv2.ellipse(img, (250, 200), (80, 80), 0, 0, 180, (200, 200, 200), -1)  # Yarım daire (dolgulu)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('cv2.line() ve cv2.ellipse() Örnekleri', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 4. Metin Ekleme: `cv2.putText()`

Kullanılabilir fontlar:
- `cv2.FONT_HERSHEY_SIMPLEX` — Sade
- `cv2.FONT_HERSHEY_COMPLEX` — Karmaşık
- `cv2.FONT_HERSHEY_DUPLEX` — Çift çizgi
- `cv2.FONT_HERSHEY_TRIPLEX` — Üç çizgi
- `cv2.FONT_HERSHEY_SCRIPT_SIMPLEX` — El yazısı
- `cv2.FONT_ITALIC` — İtalik (diğerleriyle birleştirilebilir)

```python
img = np.zeros((600, 700, 3), dtype=np.uint8)
img[:] = [30, 30, 30]

fontlar = [
    (cv2.FONT_HERSHEY_SIMPLEX, 'SIMPLEX'),
    (cv2.FONT_HERSHEY_COMPLEX, 'COMPLEX'),
    (cv2.FONT_HERSHEY_DUPLEX, 'DUPLEX'),
    (cv2.FONT_HERSHEY_TRIPLEX, 'TRIPLEX'),
    (cv2.FONT_HERSHEY_SCRIPT_SIMPLEX, 'SCRIPT_SIMPLEX'),
    (cv2.FONT_HERSHEY_SCRIPT_COMPLEX, 'SCRIPT_COMPLEX'),
    (cv2.FONT_HERSHEY_SIMPLEX | cv2.FONT_ITALIC, 'SIMPLEX + ITALIC'),
]

renkler = [
    (0, 255, 0), (255, 255, 0), (0, 255, 255),
    (255, 0, 255), (0, 165, 255), (255, 100, 100),
    (200, 200, 200)
]

for i, ((font, isim), renk) in enumerate(zip(fontlar, renkler)):
    y = 70 + i * 75
    cv2.putText(img, f'{isim}: Merhaba!', (30, y), font, 1.0, renk, 2)

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('OpenCV Font Turleri', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Farklı boyutlarda metin
img = np.zeros((400, 600, 3), dtype=np.uint8)
img[:] = [30, 30, 30]

boyutlar = [0.5, 1.0, 1.5, 2.0, 2.5]
for i, boyut in enumerate(boyutlar):
    y = 50 + i * 70
    cv2.putText(img, f'Boyut: {boyut}', (30, y),
                cv2.FONT_HERSHEY_SIMPLEX, boyut, (255, 255, 255), 2)

plt.figure(figsize=(9, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Farklı Metin Boyutları', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 5. Kontur Algılama

Kontur, bir nesnenin sınır çizgisidir. `cv2.findContours()` ile bulunur.

**Gereksinimler:**
- Girdi ikili (binary) görüntü olmalı
- Beyaz nesneler, siyah arka plan

```python
# Şekiller içeren görüntü oluştur
img_sekiller = np.zeros((500, 600, 3), dtype=np.uint8)

# Farklı şekiller çiz
cv2.rectangle(img_sekiller, (30, 30), (150, 130), (255, 255, 255), -1)    # Kare
cv2.circle(img_sekiller, (300, 80), 60, (255, 255, 255), -1)              # Daire
cv2.ellipse(img_sekiller, (500, 80), (60, 40), 0, 0, 360, (255, 255, 255), -1)  # Elips

# Üçgen
ucgen = np.array([[100, 350], [200, 200], [300, 350]], np.int32)
cv2.fillPoly(img_sekiller, [ucgen], (255, 255, 255))

# Beşgen
besgen = np.array([[450, 220], [530, 260], [510, 350], [390, 350], [370, 260]], np.int32)
cv2.fillPoly(img_sekiller, [besgen], (255, 255, 255))

# Küçük dikdörtgen
cv2.rectangle(img_sekiller, (30, 400), (120, 470), (255, 255, 255), -1)

# Gri tonlamaya çevir ve eşikle
gray_sekiller = cv2.cvtColor(img_sekiller, cv2.COLOR_BGR2GRAY)
_, thresh_sekiller = cv2.threshold(gray_sekiller, 127, 255, cv2.THRESH_BINARY)

# Konturları bul
contours, hierarchy = cv2.findContours(
    thresh_sekiller, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
)

print(f"Bulunan kontur sayısı: {len(contours)}")

# Konturları çiz
img_konturlu = img_sekiller.copy()
cv2.drawContours(img_konturlu, contours, -1, (0, 255, 0), 3)

goster(
    ['Orijinal Şekiller', 'İkili Görüntü', 'Algılanan Konturlar'],
    [img_sekiller, thresh_sekiller, img_konturlu]
)
```

## 6. Kontur Özellikleri

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Her konturun özelliklerini incele
img_ozellik = img_sekiller.copy()

print(f"{'No':<4} {'Alan':>10} {'Çevre':>10} {'Köşe Sayısı':>14} {'Şekil Tahmini':<18}")
print("-" * 60)

for i, cnt in enumerate(contours):
    # Alan
    alan = cv2.contourArea(cnt)
    
    # Çevre
    cevre = cv2.arcLength(cnt, True)
    
    # Yaklaşık poligon (köşe sayısı)
    epsilon = 0.04 * cevre
    approx = cv2.approxPolyDP(cnt, epsilon, True)
    kose_sayisi = len(approx)
    
    # Şekil tahmini
    if kose_sayisi == 3:
        sekil = "Üçgen"
    elif kose_sayisi == 4:
        # Kare mi dikdörtgen mi?
        x, y, w, h = cv2.boundingRect(cnt)
        oran = w / float(h)
        sekil = "Kare" if 0.85 <= oran <= 1.15 else "Dikdörtgen"
    elif kose_sayisi == 5:
        sekil = "Beşgen"
    elif kose_sayisi > 6:
        sekil = "Daire/Elips"
    else:
        sekil = f"Çokgen ({kose_sayisi})"
    
    # Bounding rectangle çiz
    x, y, w, h = cv2.boundingRect(cnt)
    cv2.rectangle(img_ozellik, (x, y), (x + w, y + h), (0, 255, 255), 2)
    
    # Ağırlık merkezi
    M = cv2.moments(cnt)
    if M['m00'] != 0:
        cx = int(M['m10'] / M['m00'])
        cy = int(M['m01'] / M['m00'])
        cv2.circle(img_ozellik, (cx, cy), 5, (0, 0, 255), -1)
        cv2.putText(img_ozellik, sekil, (cx - 30, cy - 15),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    
    print(f"{i+1:<4} {alan:>10.0f} {cevre:>10.1f} {kose_sayisi:>14} {sekil:<18}")

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(img_ozellik, cv2.COLOR_BGR2RGB))
plt.title('Kontur Özellikleri: Bounding Box ve Şekil Tahmini', fontsize=13, fontweight='bold')
plt.axis('off')
plt.show()
```

## 7. Pratik: Görüntüdeki Şekilleri Say ve Sınıfla

```python
# Karmaşık bir sahne oluştur
np.random.seed(42)
sahne = np.zeros((600, 800, 3), dtype=np.uint8)
sahne[:] = [20, 20, 20]

sekil_sayaci = {'Daire': 0, 'Kare': 0, 'Dikdörtgen': 0, 'Üçgen': 0}

# Rastgele daireler
for _ in range(5):
    cx = np.random.randint(50, 750)
    cy = np.random.randint(50, 550)
    r = np.random.randint(20, 50)
    cv2.circle(sahne, (cx, cy), r, (255, 255, 255), -1)
    sekil_sayaci['Daire'] += 1

# Rastgele kareler
for _ in range(4):
    x = np.random.randint(50, 700)
    y = np.random.randint(50, 500)
    boyut = np.random.randint(30, 70)
    cv2.rectangle(sahne, (x, y), (x + boyut, y + boyut), (255, 255, 255), -1)
    sekil_sayaci['Kare'] += 1

# Rastgele dikdörtgenler
for _ in range(3):
    x = np.random.randint(50, 650)
    y = np.random.randint(50, 450)
    w = np.random.randint(60, 120)
    h = np.random.randint(30, 50)
    cv2.rectangle(sahne, (x, y), (x + w, y + h), (255, 255, 255), -1)
    sekil_sayaci['Dikdörtgen'] += 1

# Rastgele üçgenler
for _ in range(3):
    cx = np.random.randint(100, 700)
    cy = np.random.randint(100, 500)
    boyut = np.random.randint(30, 60)
    pts = np.array([
        [cx, cy - boyut],
        [cx - boyut, cy + boyut],
        [cx + boyut, cy + boyut]
    ], np.int32)
    cv2.fillPoly(sahne, [pts], (255, 255, 255))
    sekil_sayaci['Üçgen'] += 1

print("Oluşturulan şekiller (gerçek):")
for sekil, sayi in sekil_sayaci.items():
    print(f"  {sekil}: {sayi}")
print(f"  Toplam: {sum(sekil_sayaci.values())}")

plt.figure(figsize=(10, 7))
plt.imshow(cv2.cvtColor(sahne, cv2.COLOR_BGR2RGB))
plt.title('Şekil Sayma Problemi', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Otomatik şekil algılama ve sayma
gray_sahne = cv2.cvtColor(sahne, cv2.COLOR_BGR2GRAY)
_, thresh_sahne = cv2.threshold(gray_sahne, 127, 255, cv2.THRESH_BINARY)

contours_sahne, _ = cv2.findContours(
    thresh_sahne, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
)

# Sonuç görüntüsü
sonuc = sahne.copy()
renk_harita = {
    'Daire': (0, 255, 0),
    'Kare': (255, 0, 0),
    'Dikdörtgen': (0, 0, 255),
    'Üçgen': (0, 255, 255)
}

algilanan = {'Daire': 0, 'Kare': 0, 'Dikdörtgen': 0, 'Üçgen': 0, 'Diğer': 0}

for cnt in contours_sahne:
    alan = cv2.contourArea(cnt)
    if alan < 200:  # Çok küçük nesneleri atla
        continue
    
    cevre = cv2.arcLength(cnt, True)
    epsilon = 0.04 * cevre
    approx = cv2.approxPolyDP(cnt, epsilon, True)
    kose = len(approx)
    
    # Şekil tahmini
    if kose == 3:
        sekil = 'Üçgen'
    elif kose == 4:
        x, y, w, h = cv2.boundingRect(cnt)
        oran = w / float(h)
        sekil = 'Kare' if 0.8 <= oran <= 1.2 else 'Dikdörtgen'
    elif kose > 6:
        sekil = 'Daire'
    else:
        sekil = 'Diğer'
    
    algilanan[sekil] = algilanan.get(sekil, 0) + 1
    renk = renk_harita.get(sekil, (128, 128, 128))
    
    # Konturu çiz
    cv2.drawContours(sonuc, [cnt], -1, renk, 3)
    
    # Etiket ekle
    M = cv2.moments(cnt)
    if M['m00'] != 0:
        cx = int(M['m10'] / M['m00'])
        cy = int(M['m01'] / M['m00'])
        cv2.putText(sonuc, sekil, (cx - 20, cy - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.45, renk, 2)

# Sonuçları yazdır
print("\n=== Algılama Sonuçları ===")
toplam = 0
for sekil, sayi in algilanan.items():
    if sayi > 0:
        print(f"  {sekil}: {sayi}")
        toplam += sayi
print(f"  Toplam algılanan: {toplam}")

plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(sonuc, cv2.COLOR_BGR2RGB))
plt.title(f'Algılanan Şekiller (Toplam: {toplam})', fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Sonuç karşılaştırma grafiği
kategoriler = ['Daire', 'Kare', 'Dikdörtgen', 'Üçgen']
gercek_degerler = [sekil_sayaci[k] for k in kategoriler]
algilanan_degerler = [algilanan.get(k, 0) for k in kategoriler]

x = np.arange(len(kategoriler))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
bars1 = ax.bar(x - width/2, gercek_degerler, width, label='Gerçek', color='#3498db', alpha=0.8)
bars2 = ax.bar(x + width/2, algilanan_degerler, width, label='Algılanan', color='#e74c3c', alpha=0.8)

ax.set_title('Şekil Algılama Karşılaştırması', fontsize=14, fontweight='bold')
ax.set_xlabel('Şekil Türü', fontsize=12)
ax.set_ylabel('Sayı', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(kategoriler, fontsize=11)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, axis='y')

for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.1,
            f'{int(bar.get_height())}', ha='center', va='bottom', fontsize=11)
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.1,
            f'{int(bar.get_height())}', ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()
```
