## 1. Değişkenler ve Veri Tipleri

Python'da değişken tanımlamak için herhangi bir anahtar kelimeye gerek yoktur. Değişken adı yazılır, `=` ile değer atanır.

Temel veri tipleri:
- **`str`** — Metin (karakter dizisi)
- **`int`** — Tam sayı
- **`float`** — Ondalıklı sayı
- **`bool`** — Mantıksal değer (`True` / `False`)

```python
# Değişken tanımlama örnekleri
isim = "Ayşe"          # str
yas = 25                # int
boy = 1.68              # float
ogrenci_mi = True       # bool

print(f"İsim: {isim}, Tipi: {type(isim)}")
print(f"Yaş: {yas}, Tipi: {type(yas)}")
print(f"Boy: {boy}, Tipi: {type(boy)}")
print(f"Öğrenci mi: {ogrenci_mi}, Tipi: {type(ogrenci_mi)}")
```

### Tip dönüşümleri

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Tip dönüşümleri
sayi_metin = "42"
sayi_int = int(sayi_metin)
sayi_float = float(sayi_metin)

print(f"Metin: '{sayi_metin}' → int: {sayi_int} → float: {sayi_float}")
print(f"int → str: {str(sayi_int)}")
print(f"float → int: {int(3.7)}")
print(f"bool → int: {int(True)}, {int(False)}")
```

## 2. Operatörler

### 2.1 Aritmetik Operatörler

```python
a = 17
b = 5

print(f"Toplama:       {a} + {b} = {a + b}")
print(f"Çıkarma:       {a} - {b} = {a - b}")
print(f"Çarpma:        {a} * {b} = {a * b}")
print(f"Bölme:         {a} / {b} = {a / b}")
print(f"Tam bölme:     {a} // {b} = {a // b}")
print(f"Mod (kalan):   {a} % {b} = {a % b}")
print(f"Üs alma:       {a} ** {b} = {a ** b}")
```

### 2.2 Karşılaştırma Operatörleri

```python
x = 10
y = 20

print(f"{x} == {y} → {x == y}")   # Eşit mi?
print(f"{x} != {y} → {x != y}")   # Eşit değil mi?
print(f"{x} > {y}  → {x > y}")    # Büyük mü?
print(f"{x} < {y}  → {x < y}")    # Küçük mü?
print(f"{x} >= {y} → {x >= y}")   # Büyük veya eşit mi?
print(f"{x} <= {y} → {x <= y}")   # Küçük veya eşit mi?
```

### 2.3 Mantıksal Operatörler

```python
a = True
b = False

print(f"{a} and {b} → {a and b}")   # VE: ikisi de True olmalı
print(f"{a} or {b}  → {a or b}")    # VEYA: en az biri True olmalı
print(f"not {a}     → {not a}")     # DEĞİL: tersini alır
print(f"not {b}     → {not b}")

# Pratik örnek
yas = 22
ogrenci = True
indirim = yas < 25 and ogrenci
print(f"\nYaş: {yas}, Öğrenci: {ogrenci} → İndirim hakkı: {indirim}")
```

## 3. Veri Yapıları

### 3.1 Liste (`list`)
Sıralı, değiştirilebilir, tekrarlı elemanlara izin verir.

```python
meyveler = ["elma", "armut", "çilek", "muz", "portakal"]

print("Tüm liste:", meyveler)
print("İlk eleman:", meyveler[0])
print("Son eleman:", meyveler[-1])
print("Dilimleme [1:3]:", meyveler[1:3])

# Eleman ekleme ve çıkarma
meyveler.append("kivi")
print("append sonrası:", meyveler)

meyveler.remove("armut")
print("remove sonrası:", meyveler)

print("Eleman sayısı:", len(meyveler))
print("'çilek' listede mi?", "çilek" in meyveler)
```

### 3.2 Demet (`tuple`)
Sıralı, **değiştirilemez** (immutable). Sabit veriler için kullanılır.

```python
koordinat = (41.0082, 28.9784)  # İstanbul
gunler = ("Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma")

print("Koordinat:", koordinat)
print("Enlem:", koordinat[0])
print("Boylam:", koordinat[1])
print("İş günleri:", gunler)
print("Eleman sayısı:", len(gunler))

# Demet çözme (unpacking)
enlem, boylam = koordinat
print(f"Enlem: {enlem}, Boylam: {boylam}")
```

### 3.3 Sözlük (`dict`)
Anahtar-değer çiftleri. Her anahtara karşılık bir değer gelir.

```python
ogrenci = {
    "ad": "Mehmet",
    "soyad": "Yılmaz",
    "yas": 21,
    "bolum": "Bilgisayar Mühendisliği",
    "notlar": [85, 92, 78, 95]
}

print("Ad:", ogrenci["ad"])
print("Bölüm:", ogrenci["bolum"])
print("Notlar:", ogrenci["notlar"])
print("Not ortalaması:", sum(ogrenci["notlar"]) / len(ogrenci["notlar"]))

# Yeni anahtar ekleme
ogrenci["email"] = "mehmet@ornek.com"

# Anahtarlar ve değerler
print("\nTüm anahtarlar:", list(ogrenci.keys()))
print("Tüm değerler:", list(ogrenci.values()))
```

### 3.4 Küme (`set`)
Sırasız, tekrarsız elemanlar. Matematiksel küme işlemleri yapılabilir.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("A kümesi:", a)
print("B kümesi:", b)
print("Birleşim (A | B):", a | b)
print("Kesişim (A & B):", a & b)
print("Fark (A - B):", a - b)

# Tekrarlı elemanları kaldırma
sayilar = [1, 2, 2, 3, 3, 3, 4, 4, 5]
benzersiz = set(sayilar)
print(f"\nOrijinal: {sayilar}")
print(f"Benzersiz: {benzersiz}")
```

## 4. Döngüler

### 4.1 `for` Döngüsü

### Liste üzerinde döngü

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Liste üzerinde döngü
sehirler = ["İstanbul", "Ankara", "İzmir", "Antalya", "Bursa"]

for sehir in sehirler:
    print(f"Hoş geldiniz: {sehir}")
```

### range() ile döngü

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# range() ile döngü
print("1'den 5'e kadar:", list(range(1, 6)))
print("0'dan 10'a 2'şer:", list(range(0, 11, 2)))

# Toplam hesaplama
toplam = 0
for i in range(1, 101):
    toplam += i
print(f"\n1'den 100'e kadar toplam: {toplam}")
```

### enumerate() ile indeks ve değer birlikte

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# enumerate() ile indeks ve değer birlikte
dersler = ["Matematik", "Fizik", "Kimya", "Biyoloji"]

for i, ders in enumerate(dersler, start=1):
    print(f"{i}. {ders}")
```

### 4.2 `while` Döngüsü

### Geri sayım

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Geri sayım
sayac = 5
while sayac > 0:
    print(f"Geri sayım: {sayac}")
    sayac -= 1
print("Başla!")
```

### break ve continue

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# break ve continue
print("--- break örneği ---")
for i in range(1, 11):
    if i == 6:
        print("6 bulundu, döngü durduruluyor!")
        break
    print(i)

print("\n--- continue örneği ---")
for i in range(1, 11):
    if i % 2 == 0:
        continue  # Çift sayıları atla
    print(i, end=" ")
print("(sadece tek sayılar)")
```

## 5. Koşullar (if / elif / else)

```python
not_ortalamasi = 76

if not_ortalamasi >= 90:
    harf = "AA"
elif not_ortalamasi >= 80:
    harf = "BA"
elif not_ortalamasi >= 70:
    harf = "BB"
elif not_ortalamasi >= 60:
    harf = "CB"
elif not_ortalamasi >= 50:
    harf = "CC"
else:
    harf = "FF"

print(f"Not ortalaması: {not_ortalamasi} → Harf notu: {harf}")
```

### İç içe koşullar

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# İç içe koşullar
yas = 20
ehliyet_var = True

if yas >= 18:
    if ehliyet_var:
        print("Araç kullanabilirsiniz.")
    else:
        print("Yaşınız uygun ama ehliyetiniz yok.")
else:
    print(f"Araç kullanmak için {18 - yas} yıl daha beklemelisiniz.")
```

## 6. Fonksiyonlar

Fonksiyonlar, tekrar kullanılabilir kod blokları oluşturmamızı sağlar.

```python
# Temel fonksiyon
def selamla(isim):
    """Verilen isme göre selamlama mesajı döndürür."""
    return f"Merhaba, {isim}! Hoş geldiniz."

mesaj = selamla("Zeynep")
print(mesaj)
```

### Varsayılan parametre değeri

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Varsayılan parametre değeri
def us_al(taban, us=2):
    """Bir sayının üssünü hesaplar. Varsayılan üs: 2 (kare)."""
    return taban ** us

print(f"3'ün karesi: {us_al(3)}")
print(f"2'nin küpü: {us_al(2, 3)}")
print(f"5'in 4. kuvveti: {us_al(5, 4)}")
```

### Birden fazla değer döndürme

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Birden fazla değer döndürme
def istatistik_hesapla(sayilar):
    """Bir sayı listesinin temel istatistiklerini hesaplar."""
    toplam = sum(sayilar)
    ortalama = toplam / len(sayilar)
    en_kucuk = min(sayilar)
    en_buyuk = max(sayilar)
    return toplam, ortalama, en_kucuk, en_buyuk

notlar = [85, 92, 78, 95, 88, 72, 90]
t, ort, mn, mx = istatistik_hesapla(notlar)

print(f"Notlar: {notlar}")
print(f"Toplam: {t}")
print(f"Ortalama: {ort:.2f}")
print(f"En düşük: {mn}")
print(f"En yüksek: {mx}")
```

### *args ile değişken sayıda parametre

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# *args ile değişken sayıda parametre
def toplam_hesapla(*sayilar):
    """İstediğiniz kadar sayı gönderin, toplamını hesaplar."""
    return sum(sayilar)

print(toplam_hesapla(1, 2, 3))
print(toplam_hesapla(10, 20, 30, 40, 50))
```

## 7. String (Metin) Metotları

Python'da metinler üzerinde çok sayıda kullanışlı metot bulunur.

```python
metin = "  Veri Bilimi ve Yapay Zekâ Eğitimi  "

print("Orijinal:     |", metin, "|")
print("strip():      |", metin.strip(), "|")
print("lower():      ", metin.strip().lower())
print("upper():      ", metin.strip().upper())
print("title():      ", metin.strip().title())
print("replace():    ", metin.strip().replace("Eğitimi", "Kursu"))
```

### split ve join

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# split ve join
cumle = "Python,Java,C++,JavaScript,Go"
diller = cumle.split(",")
print("split sonucu:", diller)

birlesik = " | ".join(diller)
print("join sonucu:", birlesik)
```

### Arama ve kontrol metotları

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Arama ve kontrol metotları
metin = "Merhaba Dünya"

print(f"'Dünya' var mı? → {'Dünya' in metin}")
print(f"startswith('Mer') → {metin.startswith('Mer')}")
print(f"endswith('ya') → {metin.endswith('ya')}")
print(f"find('ba') → indeks: {metin.find('ba')}")
print(f"count('a') → {metin.count('a')}")
print(f"isdigit() → {'12345'.isdigit()}")
print(f"isalpha() → {'Merhaba'.isalpha()}")
```

### f-string formatlama

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# f-string formatlama
ad = "Ali"
puan = 87.456

print(f"Öğrenci: {ad}")
print(f"Puan: {puan:.2f}")          # 2 ondalık
print(f"Puan: {puan:.0f}")          # Tam sayı
print(f"Yüzde: {puan/100:.1%}")     # Yüzde formatı
print(f"{'Sağa hizalı':>20}")       # Sağa hizalama
print(f"{'Sola hizalı':<20}|")
print(f"{'Ortalı':^20}")            # Ortaya hizalama
```
