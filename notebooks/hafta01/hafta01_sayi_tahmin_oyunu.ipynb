# Hafta 1: Sayı Tahmin Oyunu

Bu defterde Python'un temel yapılarını (döngüler, koşullar, fonksiyonlar, modüller) kullanarak bir **sayı tahmin oyunu** yapacağız.

## Oyun Kuralları

1. Bilgisayar 1 ile 100 arasında rastgele bir sayı seçer.
2. Kullanıcı tahmin girer.
3. Bilgisayar "Daha büyük" veya "Daha küçük" ipucu verir.
4. Kullanıcı doğru sayıyı bulana kadar devam eder.
5. Toplam tahmin sayısı ekrana yazdırılır.

---
## Adım 1: `random` Modülü

Python'un `random` modülü, rastgele sayı üretmek için kullanılır.

```python
import random

# Farklı random fonksiyonları
print("random.randint(1, 100):", random.randint(1, 100))    # 1-100 arası tam sayı
print("random.randint(1, 100):", random.randint(1, 100))    # Her seferinde farklı
print("random.random():      ", random.random())             # 0-1 arası ondalıklı
print("random.uniform(1, 10):", random.uniform(1, 10))      # 1-10 arası ondalıklı

# Listeden rastgele seçim
renkler = ["kırmızı", "mavi", "yeşil", "sarı", "turuncu"]
print("random.choice():      ", random.choice(renkler))
```

## Adım 2: Temel Oyun Döngüsü

Bilgisayar bir sayı seçer, kullanıcı `while` döngüsü içinde tahmin eder.

```python
import random

# Bilgisayarın seçtiği sayı
gizli_sayi = random.randint(1, 100)
tahmin_sayisi = 0

print("Sayı Tahmin Oyunu")
print("1 ile 100 arasında bir sayı tuttum. Tahmin edin!")
print("-" * 40)

while True:
    tahmin_girdi = input("Tahmininiz: ")
    
    # Sayı kontrolü
    try:
        tahmin = int(tahmin_girdi)
    except ValueError:
        print("Lütfen geçerli bir sayı girin!")
        continue
    
    tahmin_sayisi += 1
    
    if tahmin < gizli_sayi:
        print("↑ Daha büyük bir sayı deneyin.")
    elif tahmin > gizli_sayi:
        print("↓ Daha küçük bir sayı deneyin.")
    else:
        print(f"\nTebrikler! Doğru cevap: {gizli_sayi}")
        print(f"Toplam tahmin sayısı: {tahmin_sayisi}")
        break
```

---
## Adım 3: İpucu Sistemi Ekleme

Oyunu daha eğlenceli hale getirelim: tahmin gizli sayıya ne kadar yakınsa o kadar sıcak, ne kadar uzaksa o kadar soğuk olsun.

```python
import random

gizli_sayi = random.randint(1, 100)
tahmin_sayisi = 0

def sicaklik_ipucu(tahmin, hedef):
    """Tahminin hedefe olan uzaklığına göre sıcaklık ipucu verir."""
    fark = abs(tahmin - hedef)
    if fark == 0:
        return "DOĞRU!"
    elif fark <= 3:
        return "Çok sıcak! Neredeyse buldunuz!"
    elif fark <= 10:
        return "Sıcak! Yaklaşıyorsunuz."
    elif fark <= 25:
        return "Ilık. Doğru yoldasınız."
    elif fark <= 50:
        return "Soğuk. Daha çok deneyin."
    else:
        return "Buz gibi! Çok uzaktasınız."

print("Sayı Tahmin Oyunu (İpuçlu Versiyon)")
print("1 ile 100 arasında bir sayı tuttum.")
print("-" * 40)

while True:
    tahmin_girdi = input("Tahmininiz: ")
    
    try:
        tahmin = int(tahmin_girdi)
    except ValueError:
        print("Lütfen geçerli bir sayı girin!")
        continue
    
    if tahmin < 1 or tahmin > 100:
        print("1 ile 100 arasında bir sayı girin!")
        continue
    
    tahmin_sayisi += 1
    ipucu = sicaklik_ipucu(tahmin, gizli_sayi)
    
    if tahmin == gizli_sayi:
        print(f"\nTebrikler! Sayı {gizli_sayi} idi.")
        print(f"Toplam {tahmin_sayisi} tahminde buldunuz!")
        
        if tahmin_sayisi <= 3:
            print("Harika! Çok yeteneklisiniz!")
        elif tahmin_sayisi <= 7:
            print("Güzel performans!")
        else:
            print("İdeal strateji ile 7 tahminde bulabilirdiniz (ikili arama).")
        break
    else:
        yon = "↑ Daha büyük" if tahmin < gizli_sayi else "↓ Daha küçük"
        print(f"  {yon} | {ipucu}")
```

## Adım 4: Fonksiyon Versiyonu

Oyunu parametreleri ayarlanabilir bir fonksiyon haline getirelim.

```python
import random

def sayi_tahmin_oyunu(alt_sinir=1, ust_sinir=100, maks_tahmin=None):
    """
    Sayı tahmin oyunu.
    
    Parametreler:
        alt_sinir (int): En küçük sayı (varsayılan: 1)
        ust_sinir (int): En büyük sayı (varsayılan: 100)
        maks_tahmin (int veya None): Maksimum tahmin hakkı (None = sınırsız)
    
    Döndürür:
        dict: Oyun sonuç bilgileri
    """
    gizli_sayi = random.randint(alt_sinir, ust_sinir)
    tahminler = []
    
    print(f"Sayı Tahmin Oyunu ({alt_sinir}-{ust_sinir})")
    if maks_tahmin:
        print(f"Tahmin hakkınız: {maks_tahmin}")
    print("=" * 40)
    
    while True:
        # Maksimum tahmin kontrolü
        if maks_tahmin and len(tahminler) >= maks_tahmin:
            print(f"\nTahmin hakkınız bitti! Doğru cevap: {gizli_sayi}")
            return {
                "kazandi": False,
                "gizli_sayi": gizli_sayi,
                "tahmin_sayisi": len(tahminler),
                "tahminler": tahminler
            }
        
        kalan = f" (Kalan: {maks_tahmin - len(tahminler)})" if maks_tahmin else ""
        tahmin_girdi = input(f"Tahmin #{len(tahminler) + 1}{kalan}: ")
        
        try:
            tahmin = int(tahmin_girdi)
        except ValueError:
            print("Geçerli bir sayı girin!")
            continue
        
        if tahmin < alt_sinir or tahmin > ust_sinir:
            print(f"{alt_sinir} ile {ust_sinir} arasında bir sayı girin!")
            continue
        
        tahminler.append(tahmin)
        
        if tahmin == gizli_sayi:
            print(f"\nTebrikler! {len(tahminler)} tahminde buldunuz!")
            print(f"Gizli sayı: {gizli_sayi}")
            print(f"Tahminleriniz: {tahminler}")
            return {
                "kazandi": True,
                "gizli_sayi": gizli_sayi,
                "tahmin_sayisi": len(tahminler),
                "tahminler": tahminler
            }
        elif tahmin < gizli_sayi:
            print("  ↑ Daha büyük!")
        else:
            print("  ↓ Daha küçük!")
```

## Adım 5: Bilgisayar Tahmin Ediyor (İkili Arama)

Şimdi rolleri değiştirelim: siz bir sayı tutun, bilgisayar tahmin etsin! Bilgisayar **ikili arama** (binary search) algoritmasını kullanır. Bu algoritma her tahminde olası sayıları yarıya indirir.

```python
def bilgisayar_tahmin_eder(alt_sinir=1, ust_sinir=100):
    """
    Bilgisayar ikili arama ile sayınızı tahmin eder.
    Kullanıcı 'b' (büyük), 'k' (küçük) veya 'd' (doğru) ile yanıt verir.
    """
    print(f"{alt_sinir} ile {ust_sinir} arasında bir sayı tutun.")
    print("Hazır olduğunuzda Enter'a basın...")
    input()
    
    alt = alt_sinir
    ust = ust_sinir
    tahmin_sayisi = 0
    
    while alt <= ust:
        tahmin = (alt + ust) // 2
        tahmin_sayisi += 1
        
        print(f"\nTahminim: {tahmin}")
        cevap = input("  (b) Daha büyük | (k) Daha küçük | (d) Doğru: ").lower().strip()
        
        if cevap == 'd':
            print(f"\nBuldum! Sayınız {tahmin}.")
            print(f"Toplam {tahmin_sayisi} tahminde buldum!")
            return tahmin_sayisi
        elif cevap == 'b':
            alt = tahmin + 1
        elif cevap == 'k':
            ust = tahmin - 1
        else:
            print("Lütfen 'b', 'k' veya 'd' girin.")
            tahmin_sayisi -= 1  # Geçersiz cevabı sayma
    
    print("Hmm, bir tutarsızlık var. Sayınızı doğru mu tutuyorsunuz?")
    return tahmin_sayisi
```

## Adım 6: Otomatik Simülasyon

İnsan girişi olmadan bilgisayarın kendi kendine oynamasını simüle edelim. Bu, ikili arama algoritmasının verimliliğini gösterir.

```python
import random
import math

def otomatik_oyun(alt_sinir=1, ust_sinir=100):
    """Bilgisayar hem sayıyı tutar hem tahmin eder (ikili arama)."""
    gizli = random.randint(alt_sinir, ust_sinir)
    alt = alt_sinir
    ust = ust_sinir
    adimlar = []
    
    while alt <= ust:
        tahmin = (alt + ust) // 2
        adimlar.append(tahmin)
        
        if tahmin == gizli:
            break
        elif tahmin < gizli:
            alt = tahmin + 1
        else:
            ust = tahmin - 1
    
    return gizli, adimlar

# 5 tur simülasyon
print(f"{'Gizli Sayı':>10} | {'Adım Sayısı':>11} | Tahminler")
print("-" * 55)

for _ in range(5):
    sayi, adimlar = otomatik_oyun(1, 100)
    print(f"{sayi:>10} | {len(adimlar):>11} | {adimlar}")

# 1000 tur ile ortalama tahmin sayısı
toplam_adim = 0
maks_adim = 0
for _ in range(1000):
    _, adimlar = otomatik_oyun(1, 100)
    toplam_adim += len(adimlar)
    maks_adim = max(maks_adim, len(adimlar))

print(f"\n1000 oyun simülasyonu (1-100 arası):")
print(f"  Ortalama tahmin sayısı: {toplam_adim / 1000:.2f}")
print(f"  Maksimum tahmin sayısı: {maks_adim}")
print(f"  Teorik maksimum (log₂ 100): {math.log2(100):.2f}")
```
