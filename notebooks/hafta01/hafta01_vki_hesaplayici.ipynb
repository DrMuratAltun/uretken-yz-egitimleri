# Hafta 1: Vücut Kitle İndeksi (VKİ) Hesaplayıcı

Bu defterde Python temellerini kullanarak bir **Vücut Kitle İndeksi (VKİ)** hesaplayıcı yapacağız.

## VKİ Nedir?

Vücut Kitle İndeksi (Body Mass Index — BMI), bir kişinin kilosunun boyuna göre uygun olup olmadığını değerlendirmek için kullanılan basit bir ölçüdür.

### Formül

$$VKİ = \frac{Ağırlık\ (kg)}{Boy\ (m)^2}$$

### Sınıflandırma Tablosu

| VKİ Aralığı | Sınıflandırma |
|:---:|:---:|
| < 18.5 | Zayıf |
| 18.5 – 24.9 | Normal |
| 25.0 – 29.9 | Fazla Kilolu |
| ≥ 30.0 | Obez |

---
## Adım 1: Kullanıcıdan Veri Alma

`input()` fonksiyonu ile kullanıcıdan boy ve kilo bilgisini alacağız. `input()` her zaman **string** döndürür, bu yüzden sayıya çevirmemiz gerekir.

```python
# Kullanıcıdan veri alma
boy_cm = input("Boyunuzu santimetre cinsinden girin (örn: 170): ")
kilo_kg = input("Kilonuzu kilogram cinsinden girin (örn: 70): ")

# String'den sayıya çevirme
boy_cm = float(boy_cm)
kilo_kg = float(kilo_kg)

print(f"\nGirilen bilgiler → Boy: {boy_cm} cm, Kilo: {kilo_kg} kg")
```

---
## Adım 2: VKİ Hesaplama

```python
# Boy'u metreye çevirme
boy_m = boy_cm / 100

# VKİ hesaplama
vki = kilo_kg / (boy_m ** 2)

print(f"Boy: {boy_cm} cm = {boy_m} m")
print(f"Kilo: {kilo_kg} kg")
print(f"VKİ: {vki:.2f}")
```

## Adım 3: if-else ile Sınıflandırma

Hesaplanan VKİ değerine göre kişiyi sınıflandıracağız.

```python
# VKİ sınıflandırması
if vki < 18.5:
    sinif = "Zayıf"
    emoji = "⚠️"
    tavsiye = "Daha fazla kalori almanız önerilir."
elif vki < 25:
    sinif = "Normal"
    emoji = "✅"
    tavsiye = "Tebrikler! Sağlıklı bir kiloda görünüyorsunuz."
elif vki < 30:
    sinif = "Fazla Kilolu"
    emoji = "⚠️"
    tavsiye = "Dengeli beslenme ve düzenli egzersiz önerilir."
else:
    sinif = "Obez"
    emoji = "🔴"
    tavsiye = "Bir sağlık uzmanına danışmanız önerilir."

print("=" * 40)
print(f"{emoji} VKİ Sonucunuz: {vki:.2f}")
print(f"Sınıflandırma: {sinif}")
print(f"Tavsiye: {tavsiye}")
print("=" * 40)
```

---
## Adım 4: Fonksiyon Olarak Yazma

Şimdi aynı işlemi tekrar kullanılabilir bir fonksiyon haline getirelim.

```python
def vki_hesapla(boy_cm, kilo_kg):
    """
    Vücut Kitle İndeksi hesaplar.
    
    Parametreler:
        boy_cm (float): Boy (santimetre)
        kilo_kg (float): Kilo (kilogram)
    
    Döndürür:
        tuple: (vki_degeri, siniflandirma)
    """
    boy_m = boy_cm / 100
    vki = kilo_kg / (boy_m ** 2)
    
    if vki < 18.5:
        sinif = "Zayıf"
    elif vki < 25:
        sinif = "Normal"
    elif vki < 30:
        sinif = "Fazla Kilolu"
    else:
        sinif = "Obez"
    
    return round(vki, 2), sinif
```

### Fonksiyonu test edelim

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Fonksiyonu test edelim
test_verileri = [
    (170, 55),
    (175, 72),
    (165, 80),
    (180, 100),
]

for boy, kilo in test_verileri:
    deger, sinif = vki_hesapla(boy, kilo)
    print(f"Boy: {boy} cm, Kilo: {kilo} kg → VKİ: {deger} ({sinif})")
```

## Adım 5: Birden Fazla Kişi İçin Hesaplama (Döngü)

Bir döngü kullanarak birden fazla kişinin VKİ değerini hesaplayalım.

```python
def coklu_vki_hesapla():
    """
    Kullanıcıdan birden fazla kişinin bilgisini alarak VKİ hesaplar.
    'q' girildiğinde durur ve özet tablo gösterir.
    """
    sonuclar = []
    kisi_no = 1
    
    print("VKİ Hesaplayıcı — Çoklu Kişi Modu")
    print("Çıkmak için boy alanına 'q' yazın.")
    print("-" * 40)
    
    while True:
        print(f"\nKişi {kisi_no}:")
        boy_girdi = input("  Boy (cm): ")
        
        if boy_girdi.lower() == 'q':
            break
        
        kilo_girdi = input("  Kilo (kg): ")
        isim_girdi = input("  İsim (opsiyonel, boş bırakabilirsiniz): ")
        
        try:
            boy = float(boy_girdi)
            kilo = float(kilo_girdi)
            isim = isim_girdi.strip() if isim_girdi.strip() else f"Kişi {kisi_no}"
            
            deger, sinif = vki_hesapla(boy, kilo)
            sonuclar.append({
                "isim": isim,
                "boy": boy,
                "kilo": kilo,
                "vki": deger,
                "sinif": sinif
            })
            print(f"  → VKİ: {deger} ({sinif})")
            kisi_no += 1
            
        except ValueError:
            print("  Hata: Lütfen geçerli bir sayı girin!")
    
    # Özet tablo
    if sonuclar:
        print("\n" + "=" * 55)
        print(f"{'İsim':<15} {'Boy':>6} {'Kilo':>6} {'VKİ':>7} {'Sınıf':<15}")
        print("-" * 55)
        for s in sonuclar:
            print(f"{s['isim']:<15} {s['boy']:>5.0f}  {s['kilo']:>5.1f}  {s['vki']:>6.2f}  {s['sinif']:<15}")
        print("=" * 55)
        
        # İstatistikler
        vki_degerleri = [s["vki"] for s in sonuclar]
        print(f"\nToplam kişi: {len(sonuclar)}")
        print(f"Ortalama VKİ: {sum(vki_degerleri) / len(vki_degerleri):.2f}")
        print(f"En düşük VKİ: {min(vki_degerleri):.2f}")
        print(f"En yüksek VKİ: {max(vki_degerleri):.2f}")
    else:
        print("\nHiç veri girilmedi.")
```
