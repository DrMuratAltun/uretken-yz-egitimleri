## 1. Vaka 1: Hayali Hukuk Davası

Bir avukat ChatGPT'ye dava emsali sordu, sistem 6 sahte dava uydurdu, avukat kullandı. Hakim 5000 USD ceza verdi.

**Soru:** Bu avukatın yapması gereken neydi? Bir kontrol listesi yaz.


## 2. Vaka 2: Deepfake Politikacı

Sosyal medyada bir politikacının gerçekte söylemediği şeyler söylediği video yayıldı. Seçmen davranışı etkilendi.

**Soru:** Platformlar ne yapmalı? KVKK kapsamında nasıl şikayet edersin?


## 3. Vaka 3: Öğrenci Ödevi

Lise öğrencisi tarih ödevini ChatGPT ile yazdırdı, atıf yapmadı. Öğretmen GPTZero ile tespit etti.

**Soru:** Sınıf protokolün ne olmalı? "Yardım aldı" ile "kopya çekti" arasındaki sınır?


## 4. Vaka 4: Görsel Telif

Tasarımcı Midjourney ile poster yaptı, müşteriye sattı. Sonra başka bir sanatçının "stilini" taklit ettiği iddia edildi.

**Soru:** "Stil" telifli mi? Müşteriye ne diyeceksin?


## 5. Vaka 5: KVKK ve Hasta Verisi

Doktor, ChatGPT'ye hasta vakası yazdı, hastanın adını da ekledi. KVKK ihlali oluştu.

**Soru:** Lokal LLM neden burada zorunlu? Hasta verisi YZ'ye nasıl verilir?
