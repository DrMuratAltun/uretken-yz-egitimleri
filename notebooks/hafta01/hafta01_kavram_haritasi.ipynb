## 1. YZ vs ML vs DL vs Üretken YZ

En geniş kümeden en dar olanına doğru ilerleyen 4 katmanlı bir set düşün:

- **YZ (Yapay Zekâ):** İnsan zekâsını taklit eden tüm sistemler (sembolik AI, kural tabanlı dahil).
- **ML (Makine Öğrenmesi):** Veriden örüntü çıkaran YZ alt kümesi.
- **DL (Derin Öğrenme):** Çok katmanlı sinir ağları kullanan ML alt kümesi.
- **Üretken YZ:** İçerik (metin, görsel, ses, video) üreten modeller. Genelde DL temellidir.


> 💡 **İpucu:** Üretken YZ ≠ ML'in karşıtı. Üretken YZ, ML/DL'in bir uygulama tipi.


## 2. Diskriminatif vs Üretken Model

İki temel ML görev türü vardır:

- **Diskriminatif (sınıflandırıcı):** Veri verildiğinde etiket tahmin eder. Örnek: "Bu mail spam mi?"
- **Üretken:** Veri dağılımını öğrenir, yeni örnek üretir. Örnek: "Spam-benzeri yeni bir mail metni yaz."


> 💡 **İpucu:** ChatGPT, Gemini, DALL-E, Sora hep üretken. Yorum analizi, OCR, yüz tanıma diskriminatif.
