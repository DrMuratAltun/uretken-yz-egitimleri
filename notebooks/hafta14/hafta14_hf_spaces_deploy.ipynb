## 1. 1. Space Aç

huggingface.co → New Space → Streamlit/Gradio seç → MIT lisansı.


## 2. 2. Dosyaları Yükle

Web arayüzünden (drag-drop) ya da git ile:
```bash
git clone https://huggingface.co/spaces/USER/SPACENAME
cp app.py SPACENAME/
cp requirements.txt SPACENAME/
cd SPACENAME && git add . && git commit -m "ilk" && git push
```


## 3. 3. Secrets

Space → Settings → Repository secrets → API_KEY ekle. Kod'da `os.environ["GEMINI_API_KEY"]` ile oku.


## 4. 4. Auto-Build

Push edildi → otomatik build → 1-3 dakika → canlı URL.


## 5. 5. Custom Domain (opsiyonel)

Pro plan ile özel domain. Free'de huggingface.co/spaces/USER/SPACE URL'i.
