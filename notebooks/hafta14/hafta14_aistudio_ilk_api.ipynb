## Güvenli Saklama (.env)




```python
# .env dosyası (kodla aynı klasörde):

# GEMINI_API_KEY=AIzaSy...



# Notebook'a yükle:

from dotenv import load_dotenv

import os

load_dotenv()



API_KEY = os.environ["GEMINI_API_KEY"]

print("Key okundu, uzunluk:", len(API_KEY))

```

## İlk Çağrı




```python
import google.generativeai as genai



genai.configure(api_key=API_KEY)

model = genai.GenerativeModel("gemini-2.5-flash")



cevap = model.generate_content("Türkiye'nin başkenti?")

print(cevap.text)

```

## Streaming




```python
stream = model.generate_content(

    "Üretken YZ'yi 3 paragrafla anlat.",

    stream=True,

)

for chunk in stream:

    print(chunk.text, end="", flush=True)

```

## System Message




```python
model = genai.GenerativeModel(

    "gemini-2.5-flash",

    system_instruction="Sen 20 yıllık bir Türk matematik öğretmenisin. Türkçe konuşursun.",

)



r = model.generate_content("Üslü sayıyı 5. sınıfa nasıl anlatırsın?")

print(r.text)

```

## Yapılandırılmış Çıktı (JSON)




```python
import json



response = model.generate_content(

    'Bu yorumdan duygu, ürün, şikayet kategorisi çıkar. JSON: {"duygu": "POZ|NEG|NÖT", "urun": "...", "kategori": "kargo|kalite|fiyat|diğer"}\n\nYorum: "Ürün güzel ama kargo geç geldi."',

    generation_config={"response_mime_type": "application/json"}

)

data = json.loads(response.text)

print(data)

```
