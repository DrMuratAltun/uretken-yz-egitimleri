## app.py — Tüm Kod




```python
# Bu kodu app.py adıyla kaydet

import streamlit as st

import google.generativeai as genai

import os



st.set_page_config(page_title="Kişisel YZ Asistanım", page_icon="✨")

st.title("✨ Kişisel YZ Asistanım")

st.caption("Üretken YZ Atölyesi · Capstone şablonu")



# API key (Streamlit Secrets ya da env)

genai.configure(api_key=st.secrets.get("GEMINI_API_KEY", os.environ.get("GEMINI_API_KEY")))



# Sidebar — kişiselleştirme

with st.sidebar:

    st.header("Ayarlar")

    rol = st.text_area("Rol", "Sen yardımsever bir Türk eğitim danışmanısın.")

    sicaklik = st.slider("Yaratıcılık", 0.0, 1.5, 0.7)



model = genai.GenerativeModel("gemini-2.5-flash", system_instruction=rol)



# Chat geçmişi

if "messages" not in st.session_state:

    st.session_state.messages = []



for msg in st.session_state.messages:

    with st.chat_message(msg["role"]):

        st.write(msg["content"])



if prompt := st.chat_input("Sorunu yaz..."):

    st.session_state.messages.append({"role": "user", "content": prompt})

    with st.chat_message("user"):

        st.write(prompt)



    with st.chat_message("assistant"):

        with st.spinner("Düşünüyor..."):

            r = model.generate_content(

                prompt,

                generation_config={"temperature": sicaklik},

            )

            st.write(r.text)

            st.session_state.messages.append({"role": "assistant", "content": r.text})

```
