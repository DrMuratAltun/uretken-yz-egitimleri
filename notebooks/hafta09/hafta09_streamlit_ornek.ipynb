## 2. Önce Modeli Hazırlayalım

Streamlit uygulamasında kullanmak üzere bir model eğitip kaydedelim.

````python
import numpy as np
import joblib
import os
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

# Veri yükle ve eğit
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))
])
pipeline.fit(X_train, y_train)

accuracy = accuracy_score(y_test, pipeline.predict(X_test))
print(f"Model doğruluğu: {accuracy:.4f}")

# Modeli kaydet
os.makedirs('models', exist_ok=True)
joblib.dump(pipeline, 'models/iris_pipeline.joblib')
print("Model kaydedildi: models/iris_pipeline.joblib")
````

## 4. Streamlit Uygulama Dosyası Oluşturma

`%%writefile` komutu ile Python dosyası oluşturuyoruz. Bu dosya Streamlit ile çalıştırılacak.

````python
%%writefile app.py
import streamlit as st
import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt

# Sayfa ayarları
st.set_page_config(
    page_title="Iris Çiçeği Sınıflandırıcı",
    page_icon="🌸",
    layout="wide"
)

# Başlık
st.title("🌸 Iris Çiçeği Sınıflandırıcı")
st.markdown("Bu uygulama, çiçek ölçümlerine göre Iris türünü tahmin eder.")

# Model yükle
@st.cache_resource
def load_model():
    return joblib.load('models/iris_pipeline.joblib')

model = load_model()

# Yan panel (Sidebar)
st.sidebar.title("📊 Parametre Ayarları")
st.sidebar.markdown("Çiçek ölçümlerini aşağıdan ayarlayın:")

sepal_length = st.sidebar.slider(
    "Çanak Yaprak Uzunluğu (cm)",
    min_value=4.0, max_value=8.0, value=5.4, step=0.1
)
sepal_width = st.sidebar.slider(
    "Çanak Yaprak Genişliği (cm)",
    min_value=2.0, max_value=4.5, value=3.4, step=0.1
)
petal_length = st.sidebar.slider(
    "Taç Yaprak Uzunluğu (cm)",
    min_value=1.0, max_value=7.0, value=4.7, step=0.1
)
petal_width = st.sidebar.slider(
    "Taç Yaprak Genişliği (cm)",
    min_value=0.1, max_value=2.5, value=1.4, step=0.1
)

# Seçili tür filtresi
st.sidebar.markdown("---")
gosterim_modu = st.sidebar.selectbox(
    "Görüntüleme Modu",
    ["Tahmin Sonucu", "Veri Seti Karşılaştırma", "Model Bilgisi"]
)

# Ana içerik
col1, col2 = st.columns(2)

with col1:
    st.subheader("📝 Girilen Ölçümler")
    input_data = pd.DataFrame({
        'Özellik': ['Çanak Yaprak Uzunluğu', 'Çanak Yaprak Genişliği',
                    'Taç Yaprak Uzunluğu', 'Taç Yaprak Genişliği'],
        'Değer (cm)': [sepal_length, sepal_width, petal_length, petal_width]
    })
    st.dataframe(input_data, use_container_width=True, hide_index=True)

with col2:
    st.subheader("🎯 Tahmin Sonucu")
    
    features = np.array([[sepal_length, sepal_width, petal_length, petal_width]])
    prediction = model.predict(features)[0]
    probabilities = model.predict_proba(features)[0]
    
    species_names = ['Setosa', 'Versicolor', 'Virginica']
    species_emojis = ['🌼', '🌺', '🌷']
    
    predicted_species = species_names[prediction]
    confidence = probabilities[prediction] * 100
    
    st.metric(
        label="Tahmin Edilen Tür",
        value=f"{species_emojis[prediction]} {predicted_species}",
        delta=f"Güven: %{confidence:.1f}"
    )

# Olasılık grafiği
st.subheader("📊 Sınıf Olasılıkları")

fig, ax = plt.subplots(figsize=(8, 3))
colors = ['#3498db', '#e74c3c', '#2ecc71']
bars = ax.barh(species_names, probabilities * 100, color=colors, edgecolor='white', height=0.5)
ax.set_xlabel('Olasılık (%)')
ax.set_xlim(0, 105)
for bar, prob in zip(bars, probabilities):
    ax.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
            f'%{prob*100:.1f}', va='center', fontsize=12)
plt.tight_layout()
st.pyplot(fig)

# Görüntüleme moduna göre ek bilgi
if gosterim_modu == "Veri Seti Karşılaştırma":
    st.subheader("📈 Veri Seti ile Karşılaştırma")
    from sklearn.datasets import load_iris
    iris = load_iris()
    df_iris = pd.DataFrame(iris.data, columns=['Çanak Uzunluk', 'Çanak Genişlik',
                                                'Taç Uzunluk', 'Taç Genişlik'])
    df_iris['Tür'] = [species_names[t] for t in iris.target]
    
    fig2, ax2 = plt.subplots(figsize=(10, 6))
    for i, species in enumerate(species_names):
        subset = df_iris[df_iris['Tür'] == species]
        ax2.scatter(subset['Taç Uzunluk'], subset['Taç Genişlik'],
                   label=species, alpha=0.6, color=colors[i], s=50)
    ax2.scatter(petal_length, petal_width, c='black', marker='X', s=200,
               zorder=5, label='Sizin Girdiniz')
    ax2.set_xlabel('Taç Yaprak Uzunluğu (cm)')
    ax2.set_ylabel('Taç Yaprak Genişliği (cm)')
    ax2.set_title('Iris Veri Seti Dağılımı')
    ax2.legend()
    st.pyplot(fig2)

elif gosterim_modu == "Model Bilgisi":
    st.subheader("ℹ️ Model Bilgisi")
    st.markdown("""
    | Özellik | Değer |
    |---------|-------|
    | **Algoritma** | Random Forest (100 ağaç) |
    | **Ön İşleme** | StandardScaler |
    | **Eğitim Verisi** | Iris Dataset (120 örnek) |
    | **Test Doğruluğu** | ~%96-100 |
    | **Özellikler** | 4 sayısal özellik |
    | **Sınıflar** | 3 (Setosa, Versicolor, Virginica) |
    """)

# Alt bilgi
st.markdown("---")
st.markdown("*Bu uygulama eğitim amaçlı oluşturulmuştur. Hafta 9 - Streamlit Örneği*")
````

## 6. Daha Gelişmiş Bir Uygulama Örneği

Aşağıda çoklu sayfa desteği ve dosya yükleme içeren gelişmiş bir örnek var.

````python
%%writefile app_advanced.py
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

st.set_page_config(page_title="Veri Analizi Aracı", layout="wide")

st.title("📊 Etkileşimli Veri Analizi Aracı")

# Dosya yükleme
uploaded_file = st.file_uploader("CSV dosyası yükleyin", type=['csv'])

if uploaded_file is not None:
    df = pd.read_csv(uploaded_file)
    
    st.subheader("Veri Önizleme")
    st.dataframe(df.head(20))
    
    st.subheader("İstatistiksel Özet")
    st.dataframe(df.describe())
    
    # Sütun seçimi
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    
    if len(numeric_cols) >= 2:
        col1, col2 = st.columns(2)
        with col1:
            x_col = st.selectbox("X Ekseni", numeric_cols)
        with col2:
            y_col = st.selectbox("Y Ekseni", numeric_cols, index=min(1, len(numeric_cols)-1))
        
        fig, ax = plt.subplots(figsize=(10, 6))
        ax.scatter(df[x_col], df[y_col], alpha=0.5, color='steelblue')
        ax.set_xlabel(x_col)
        ax.set_ylabel(y_col)
        ax.set_title(f'{x_col} vs {y_col}')
        st.pyplot(fig)
    
    # Histogram
    st.subheader("Dağılım Grafiği")
    hist_col = st.selectbox("Sütun Seçin", numeric_cols, key='hist')
    bins = st.slider("Bin Sayısı", 5, 50, 20)
    
    fig2, ax2 = plt.subplots(figsize=(10, 4))
    ax2.hist(df[hist_col].dropna(), bins=bins, color='seagreen', edgecolor='white')
    ax2.set_xlabel(hist_col)
    ax2.set_ylabel('Frekans')
    ax2.set_title(f'{hist_col} Dağılımı')
    st.pyplot(fig2)
else:
    st.info("Lütfen bir CSV dosyası yükleyin.")
    
    # Demo veri ile örnek
    if st.button("Demo Veri ile Dene"):
        demo_df = pd.DataFrame({
            'Yaş': np.random.randint(18, 65, 100),
            'Gelir': np.random.randint(3000, 25000, 100),
            'Harcama': np.random.randint(1000, 15000, 100)
        })
        st.dataframe(demo_df)
        st.line_chart(demo_df)
````
