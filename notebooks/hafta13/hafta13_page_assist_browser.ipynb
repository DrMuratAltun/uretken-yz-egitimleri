## 1. Kurulum

Chrome Web Store → "Page Assist" ara → Add to Chrome.


## 2. Ollama Bağla

Settings → Ollama URL: `http://localhost:11434`.


## 3. Kullanım

Bir web sayfası aç → Page Assist ikonuna tıkla → "Bu sayfanın özetini ver" sor. Sayfa içeriği otomatik kontekste girer.


## 4. Web Search Mode

Chat sırasında "/" tuşu → search → DuckDuckGo + Ollama ile araştırma yap.
