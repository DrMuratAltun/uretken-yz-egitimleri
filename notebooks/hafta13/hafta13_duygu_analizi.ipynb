## 1. Hugging Face Pipeline Giriş

**Hugging Face Transformers**, önceden eğitilmiş dil modellerini kolayca kullanmamızı sağlayan bir kütüphanedir.

`pipeline()` fonksiyonu, karmaşık NLP görevlerini tek satırda çözmemize olanak tanır:

| Görev | Pipeline Adı |
|-------|-------------|
| Duygu Analizi | `sentiment-analysis` |
| Metin Sınıflandırma | `text-classification` |
| Soru Cevaplama | `question-answering` |
| Metin Üretme | `text-generation` |
| Çeviri | `translation` |
| Özetleme | `summarization` |

## 2. İngilizce Duygu Analizi

### Metin İşleme ve NLP

Metin verisini makine öğrenmesi algoritmalarının anlayabileceği sayısal formata dönüştürüyoruz. Tokenizasyon, vektörizasyon ve önceden eğitilmiş modeller bu sürecin parçalarıdır.

```python
# İngilizce duygu analizi pipeline'ı oluştur
en_sentiment = pipeline("sentiment-analysis")

print("İngilizce duygu analizi modeli yüklendi!")
```

### Metin İşleme ve NLP

Metin verisini makine öğrenmesi algoritmalarının anlayabileceği sayısal formata dönüştürüyoruz. Tokenizasyon, vektörizasyon ve önceden eğitilmiş modeller bu sürecin parçalarıdır.

```python
# İngilizce örnek tweet'ler
ingilizce_tweetler = [
    "I love this new AI technology! It's amazing!",
    "This product is terrible. Worst purchase ever.",
    "The weather is nice today.",
    "I'm so happy with the results of my project!",
    "This movie was boring and waste of time.",
    "Great customer service! Really helpful team.",
    "I'm disappointed with the quality of this laptop.",
    "Machine learning is changing the world for better."
]

# Analiz
en_sonuclar = en_sentiment(ingilizce_tweetler)

print(f"{'Tweet':<55} {'Duygu':<12} {'Güven':>8}")
print("=" * 80)
for tweet, sonuc in zip(ingilizce_tweetler, en_sonuclar):
    kisaltilmis = tweet[:52] + '...' if len(tweet) > 52 else tweet
    print(f"{kisaltilmis:<55} {sonuc['label']:<12} {sonuc['score']:.4f}")
```

## 3. Türkçe Duygu Analizi

Türkçe duygu analizi için özel eğitilmiş modeller kullanacağız. `savasy/bert-base-turkish-sentiment-cased` modeli Türkçe metinlerde duygu analizi yapmak için eğitilmiştir.

```python
# Türkçe duygu analizi modeli
try:
    tr_sentiment = pipeline(
        "sentiment-analysis",
        model="savasy/bert-base-turkish-sentiment-cased"
    )
    turkce_model = "savasy/bert-base-turkish-sentiment-cased"
    print(f"Türkçe duygu analizi modeli yüklendi: {turkce_model}")
except Exception as e:
    print(f"Türkçe model yüklenemedi: {e}")
    print("Çok dilli model kullanılıyor...")
    tr_sentiment = pipeline(
        "sentiment-analysis",
        model="nlptown/bert-base-multilingual-uncased-sentiment"
    )
    turkce_model = "nlptown/bert-base-multilingual-uncased-sentiment"
    print(f"Çok dilli model yüklendi: {turkce_model}")
```

### Türkçe örnek tweet verisi

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Türkçe örnek tweet verisi
turkce_tweetler = [
    # Pozitif
    "Bu ürünü çok beğendim, harika bir kalite!",
    "Yapay zekâ alanındaki gelişmeler beni çok heyecanlandırıyor.",
    "Müşteri hizmetleri çok ilgili ve yardımsever.",
    "Bugün hava çok güzel, keyifli bir gün geçiriyorum.",
    "Bu film muhteşemdi, kesinlikle tavsiye ederim!",
    "Yeni telefonum mükemmel çalışıyor, çok memnunum.",
    "Sınav sonuçlarım çok iyi geldi, çok mutluyum!",
    # Negatif
    "Bu ürün çok kötü, paramı çöpe attım.",
    "Kargo çok geç geldi ve ürün hasarlıydı.",
    "Bu restoran berbat, yemekler soğuk ve lezzetsiz.",
    "Müşteri desteği hiç yardımcı olmadı, çok sinirli.",
    "İnternet sürekli kesiliyor, bu kabul edilemez!",
    "Bu kurs beklentilerimi hiç karşılamadı, hayal kırıklığı.",
    "Uygulama sürekli çöküyor, kullanılamaz durumda.",
    # Nötr
    "Bugün toplantı saat üçte başlayacak.",
    "Hava durumu yarın yağışlı olacak.",
    "Yeni yapay zekâ modeli duyuruldu.",
    "Kütüphane saat beşte kapanıyor.",
    "Ankara'nın nüfusu beş milyonun üzerinde.",
    "Yarınki maç saat sekizde başlıyor."
]

# Beklenen kategoriler
beklenen_kategoriler = (['Pozitif'] * 7 + ['Negatif'] * 7 + ['Nötr'] * 6)

print(f"Toplam tweet sayısı: {len(turkce_tweetler)}")
print(f"Pozitif: 7, Negatif: 7, Nötr: 6")
```

### Metin İşleme ve NLP

Metin verisini makine öğrenmesi algoritmalarının anlayabileceği sayısal formata dönüştürüyoruz. Tokenizasyon, vektörizasyon ve önceden eğitilmiş modeller bu sürecin parçalarıdır.

```python
# Türkçe tweet'leri analiz et
tr_sonuclar = tr_sentiment(turkce_tweetler)

print(f"{'Tweet':<55} {'Beklenen':<10} {'Tahmin':<12} {'Güven':>8}")
print("=" * 90)
for tweet, beklenen, sonuc in zip(turkce_tweetler, beklenen_kategoriler, tr_sonuclar):
    kisaltilmis = tweet[:52] + '...' if len(tweet) > 52 else tweet
    print(f"{kisaltilmis:<55} {beklenen:<10} {sonuc['label']:<12} {sonuc['score']:.4f}")
```

## 4. Toplu Analiz

### Sonuçları DataFrame'e dönüştür

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Sonuçları DataFrame'e dönüştür
analiz_df = pd.DataFrame({
    'Tweet': turkce_tweetler,
    'Beklenen': beklenen_kategoriler,
    'Tahmin': [s['label'] for s in tr_sonuclar],
    'Güven': [s['score'] for s in tr_sonuclar]
})

print("Analiz Sonuçları:")
print(analiz_df.to_string(index=True))
```

### Veri Gruplama ve Analiz

Verileri belirli kategorilere göre grupladıp özet istatistikler hesaplıyoruz. `groupby()` ile SQL'deki GROUP BY benzeri operasyonlar yapıyoruz.

```python
# Duygu dağılımı istatistikleri
tahmin_dagilimi = analiz_df['Tahmin'].value_counts()
beklenen_dagilimi = analiz_df['Beklenen'].value_counts()

print("Tahmin Dağılımı:")
for etiket, sayi in tahmin_dagilimi.items():
    print(f"  {etiket}: {sayi} ({sayi/len(analiz_df)*100:.1f}%)")

print(f"\nOrtalama güven skoru: {analiz_df['Güven'].mean():.4f}")
print(f"Minimum güven: {analiz_df['Güven'].min():.4f}")
print(f"Maksimum güven: {analiz_df['Güven'].max():.4f}")
```

## 5. Sonuçların Görselleştirilmesi

### Pasta Grafiği

Kategorilerin genel dağılım içindeki oranlarını pasta grafiği ile görselleştiriyoruz.

```python
# Pasta grafiği - Tahmin dağılımı
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Beklenen dağılım
renkler_beklenen = ['#2ecc71', '#e74c3c', '#3498db']
beklenen_dagilimi.plot.pie(
    ax=axes[0], autopct='%1.1f%%', colors=renkler_beklenen,
    startangle=90, textprops={'fontsize': 12}
)
axes[0].set_title('Beklenen Duygu Dağılımı', fontsize=14, fontweight='bold')
axes[0].set_ylabel('')

# Tahmin dağılımı
renkler_tahmin = []
renk_haritasi = {'positive': '#2ecc71', 'POSITIVE': '#2ecc71', 'Pozitif': '#2ecc71',
                 'negative': '#e74c3c', 'NEGATIVE': '#e74c3c', 'Negatif': '#e74c3c',
                 'neutral': '#3498db', 'NEUTRAL': '#3498db', 'Nötr': '#3498db'}
for etiket in tahmin_dagilimi.index:
    renkler_tahmin.append(renk_haritasi.get(etiket, '#95a5a6'))

tahmin_dagilimi.plot.pie(
    ax=axes[1], autopct='%1.1f%%', colors=renkler_tahmin if renkler_tahmin else None,
    startangle=90, textprops={'fontsize': 12}
)
axes[1].set_title('Model Tahmin Dağılımı', fontsize=14, fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

### Çoklu Grafik Paneli

Birden fazla grafiği yan yana veya alt alta çizdirerek karşılaştırmalı analiz yapıyoruz.

```python
# Güven skorları çubuk grafiği
fig, ax = plt.subplots(figsize=(14, 8))

renkler = []
for etiket in analiz_df['Tahmin']:
    if 'pos' in etiket.lower() or 'positive' in etiket.lower():
        renkler.append('#2ecc71')
    elif 'neg' in etiket.lower() or 'negative' in etiket.lower():
        renkler.append('#e74c3c')
    else:
        renkler.append('#3498db')

kisaltilmis_tweetler = [t[:40] + '...' if len(t) > 40 else t for t in analiz_df['Tweet']]

bars = ax.barh(range(len(analiz_df)), analiz_df['Güven'], color=renkler, edgecolor='black', alpha=0.8)
ax.set_yticks(range(len(analiz_df)))
ax.set_yticklabels(kisaltilmis_tweetler, fontsize=9)
ax.set_xlabel('Güven Skoru')
ax.set_title('Tweet Duygu Analizi - Güven Skorları', fontsize=14, fontweight='bold')
ax.set_xlim(0, 1.1)

# Güven değerlerini çubukların yanına yaz
for i, (v, etiket) in enumerate(zip(analiz_df['Güven'], analiz_df['Tahmin'])):
    ax.text(v + 0.01, i, f'{v:.3f} ({etiket})', va='center', fontsize=8)

ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

### Dağılım Görselleştirmesi

Aşağıdaki grafikte verinin dağılımını histogram ile inceliyoruz. Dağılımın şekli (normal, çarpık, bimodal) hangi istatistiksel yöntemlerin uygulanabileceğini belirler.

```python
# Güven skoru dağılımı histogramı
plt.figure(figsize=(10, 5))

for kategori in analiz_df['Beklenen'].unique():
    alt_kume = analiz_df[analiz_df['Beklenen'] == kategori]
    plt.hist(alt_kume['Güven'], bins=10, alpha=0.6, label=f'{kategori} (beklenen)', edgecolor='black')

plt.xlabel('Güven Skoru')
plt.ylabel('Frekans')
plt.title('Güven Skoru Dağılımı (Beklenen Kategorilere Göre)', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
