## Test Cümleleri




```python
# Anlamca yakın 3 çift

ciftler = [

    ("İstanbul Türkiye'nin en kalabalık şehridir.", "Türkiye'de en çok insan İstanbul'da yaşar."),

    ("Yapay zeka eğitimi yaygınlaşıyor.", "AI öğrenme programları popüler oluyor."),

    ("Bugün hava çok güzel.", "Hava yağışlı ve soğuk."),  # ← bu çift uzak olmalı

]

```

## Ollama Embedding (lokal)




```python
import ollama

import numpy as np



def cosine(a, b):

    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))



def embed_ollama(metin, model="nomic-embed-text"):

    r = ollama.embeddings(model=model, prompt=metin)

    return np.array(r["embedding"])



for c1, c2 in ciftler:

    e1, e2 = embed_ollama(c1), embed_ollama(c2)

    print(f"{cosine(e1, e2):.3f} | {c1[:40]}... ↔ {c2[:40]}...")

```

## Multilingual-e5 Karşılaştırma




```python
from sentence_transformers import SentenceTransformer



model = SentenceTransformer("intfloat/multilingual-e5-large")



def embed_e5(metin):

    return model.encode("query: " + metin)



print("\n--- multilingual-e5 ---")

for c1, c2 in ciftler:

    e1, e2 = embed_e5(c1), embed_e5(c2)

    print(f"{cosine(e1, e2):.3f} | {c1[:40]}... ↔ {c2[:40]}...")

```
