## 1. Metin Ön İşleme

NLP'de metin verisini modele vermeden önce temizleme ve düzenleme işlemleri yapılır.

### Temel Adımlar:
1. Küçük harfe dönüştürme
2. Noktalama işaretlerini kaldırma
3. Sayıları kaldırma/dönüştürme
4. Fazla boşlukları temizleme
5. Durak kelimeleri (stopwords) çıkarma

```python
# Örnek Türkçe metin
ornek_metin = """Yapay Zekâ (YZ), makinelerin insan benzeri düşünme ve öğrenme yeteneklerini 
taklit etmesini sağlayan bir bilgisayar bilimi dalıdır. 2024 yılında büyük dil modelleri 
(LLM'ler) çok popüler oldu! Derin öğrenme, makine öğrenmesi ve doğal dil işleme gibi 
alt alanları vardır. #YapayZeka @teknoloji"""

print("Orijinal metin:")
print(ornek_metin)
print(f"\nKarakter sayısı: {len(ornek_metin)}")
```

```python
def metin_temizle(metin):
    """Metin ön işleme fonksiyonu."""
    # 1. Küçük harfe dönüştür
    metin = metin.lower()
    print(f"1. Küçük harf: {metin[:80]}...")
    
    # 2. URL'leri kaldır
    metin = re.sub(r'http\S+|www\S+', '', metin)
    
    # 3. Mention ve hashtag kaldır
    metin = re.sub(r'[@#]\w+', '', metin)
    print(f"2. Mention/hashtag temizlendi: {metin[:80]}...")
    
    # 4. Noktalama işaretlerini kaldır
    metin = re.sub(r'[^\w\s]', '', metin)
    print(f"3. Noktalama temizlendi: {metin[:80]}...")
    
    # 5. Sayıları kaldır
    metin = re.sub(r'\d+', '', metin)
    print(f"4. Sayılar kaldırıldı: {metin[:80]}...")
    
    # 6. Fazla boşlukları temizle
    metin = re.sub(r'\s+', ' ', metin).strip()
    print(f"5. Boşluklar temizlendi: {metin[:80]}...")
    
    return metin

temiz_metin = metin_temizle(ornek_metin)
print(f"\nTemizlenmiş metin: {temiz_metin}")
```

## 2. NLTK ile Tokenizasyon

Tokenizasyon, metni küçük parçalara (token) ayırma işlemidir:
- **Kelime tokenizasyonu**: Metni kelimelere ayırma
- **Cümle tokenizasyonu**: Metni cümlelere ayırma

```python
from nltk.tokenize import word_tokenize, sent_tokenize

metin = """Yapay zekâ günümüzde birçok alanda kullanılmaktadır. Sağlık, eğitim ve 
finans sektörleri en çok faydalanan alanlardır. Derin öğrenme modelleri özellikle 
görüntü tanıma ve doğal dil işlemede çığır açmıştır."""

# Cümle tokenizasyonu
cumleler = sent_tokenize(metin)
print("Cümle Tokenizasyonu:")
for i, cumle in enumerate(cumleler, 1):
    print(f"  Cümle {i}: {cumle.strip()}")

print(f"\nToplam cümle sayısı: {len(cumleler)}")

# Kelime tokenizasyonu
kelimeler = word_tokenize(metin)
print(f"\nKelime Tokenizasyonu:")
print(f"  İlk 15 kelime: {kelimeler[:15]}")
print(f"  Toplam kelime sayısı: {len(kelimeler)}")
print(f"  Benzersiz kelime sayısı: {len(set(kelimeler))}")
```

## 3. Türkçe Durak Kelimeleri (Stopwords)

Durak kelimeler, anlam taşımayan sık kullanılan kelimelerdir (ve, bir, bu, ile, vb.).
Bu kelimeler genellikle NLP analizinden çıkarılır.

```python
from nltk.corpus import stopwords

# NLTK Türkçe durak kelimeler
nltk_turkce_durak = set(stopwords.words('turkish'))
print(f"NLTK Türkçe durak kelime sayısı: {len(nltk_turkce_durak)}")
print(f"Örnekler: {list(nltk_turkce_durak)[:20]}")

# Genişletilmiş Türkçe durak kelime listesi
turkce_durak_kelimeler = set([
    'acaba', 'ama', 'aslında', 'az', 'bazı', 'belki', 'belli', 'ben', 'bile',
    'bir', 'birçok', 'biri', 'birkaç', 'birkez', 'birşey', 'birşeyi', 'biz',
    'bu', 'buna', 'bunda', 'bundan', 'bunlar', 'bunları', 'bunların', 'bunu',
    'bunun', 'burada', 'çok', 'çünkü', 'da', 'daha', 'dahi', 'de', 'defa',
    'değil', 'diğer', 'diye', 'dolayı', 'eden', 'etmek', 'gibi', 'göre',
    'hala', 'hangi', 'hem', 'henüz', 'hep', 'hepsi', 'her', 'herhangi',
    'herkes', 'hiç', 'hiçbir', 'için', 'ile', 'ise', 'işte', 'kadar',
    'karşı', 'kendi', 'kez', 'ki', 'kim', 'kimse', 'mu', 'mı', 'mü',
    'nasıl', 'ne', 'neden', 'nerde', 'nere', 'nerede', 'nereye', 'niçin',
    'niye', 'o', 'olan', 'olarak', 'oldu', 'olma', 'olmak', 'olup',
    'olur', 'onu', 'onun', 'ora', 'orada', 'öyle', 'pek', 'sen', 'siz',
    'şey', 'şu', 'tüm', 'var', 've', 'veya', 'ya', 'yani', 'yerine',
    'yine', 'yoksa', 'zaten', 'zira'
])

# NLTK ve özel listeyi birleştir
tum_durak = nltk_turkce_durak.union(turkce_durak_kelimeler)
print(f"\nToplam durak kelime sayısı: {len(tum_durak)}")
```

### Durak kelimeleri çıkarma örneği

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Durak kelimeleri çıkarma örneği
ornek = "yapay zekâ bir bilgisayar bilimi dalıdır ve çok önemlidir"
kelimeler = ornek.split()

temiz_kelimeler = [k for k in kelimeler if k not in tum_durak]

print(f"Orijinal: {ornek}")
print(f"Kelimeler: {kelimeler}")
print(f"Durak kelimeler çıkarıldı: {temiz_kelimeler}")
print(f"\nÇıkarılan kelimeler: {[k for k in kelimeler if k in tum_durak]}")
```

## 4. Kök Bulma (Stemming) ve Lemmatization

### Stemming (Kök Bulma)
Kelimenin sonundaki ekleri kaldırarak köküne ulaşma. Basit kural tabanlıdır.

### Lemmatization (Sözlük Kökü Bulma)
Kelimenin sözlükteki temel formunu bulma. Daha doğru ama daha yavaştır.

**Not:** Türkçe sondan eklemeli (agglutinative) bir dil olduğu için İngilizce stemmer/lemmatizer iyi çalışmaz. Türkçe için Zemberek gibi özel araçlar önerilir.

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# İngilizce örneklerle gösterim
ingilizce_kelimeler = ['running', 'runs', 'ran', 'easily', 'fairly', 
                       'studies', 'studying', 'better', 'learning', 'learned']

print(f"{'Kelime':<15} {'Stemming':<15} {'Lemmatization':<15}")
print("-" * 45)
for kelime in ingilizce_kelimeler:
    stem = stemmer.stem(kelime)
    lemma = lemmatizer.lemmatize(kelime, pos='v')
    print(f"{kelime:<15} {stem:<15} {lemma:<15}")
```

### Basit Türkçe kök bulma fonksiyonu (eğitim amaçlı)

Aşağıdaki kod bloğunda bu işlemi gerçekleştiriyoruz.

```python
# Basit Türkçe kök bulma fonksiyonu (eğitim amaçlı)
def basit_turkce_kok(kelime):
    """Basit kural tabanlı Türkçe kök bulma (eğitim amaçlı)."""
    ekler = ['ların', 'lerin', 'ları', 'leri', 'lar', 'ler',
             'dan', 'den', 'tan', 'ten', 'nın', 'nin',
             'dır', 'dir', 'dır', 'dur', 'dür',
             'mak', 'mek', 'yor', 'iyor', 'ıyor',
             'da', 'de', 'ta', 'te']
    
    for ek in sorted(ekler, key=len, reverse=True):
        if kelime.endswith(ek) and len(kelime) - len(ek) >= 2:
            return kelime[:-len(ek)]
    return kelime

turkce_kelimeler = ['bilgisayarlar', 'öğrenmenin', 'çalışmaktadır', 'modellerden',
                    'eğitimdir', 'yapıyor', 'okuyor', 'kitapları']

print(f"{'Kelime':<20} {'Basit Kök':<15}")
print("-" * 35)
for kelime in turkce_kelimeler:
    print(f"{kelime:<20} {basit_turkce_kok(kelime):<15}")

print("\nNot: Bu basitleştirilmiş bir örnektir. Gerçek projede Zemberek kullanılmalıdır.")
```

## 5. Bag of Words (Kelime Çantası)

**Bag of Words (BoW)**, bir metni kelime frekanslarıyla temsil eden basit bir yöntemdir.
Her belge, kelime dağarcığındaki kelimelerin sayısıyla bir vektöre dönüştürülür.

```python
from sklearn.feature_extraction.text import CountVectorizer

# Türkçe belge koleksiyonu
belgeler = [
    "Yapay zekâ ve makine öğrenmesi günümüzde çok popüler",
    "Derin öğrenme yapay zekânın bir alt alanıdır",
    "Makine öğrenmesi veri bilimi için çok önemlidir",
    "Doğal dil işleme metin analizi için kullanılır",
    "Veri bilimi ve yapay zekâ birlikte çalışır"
]

# CountVectorizer ile Bag of Words
bow_vektorizer = CountVectorizer()
bow_matris = bow_vektorizer.fit_transform(belgeler)

# Sonuçları DataFrame olarak göster
kelime_dagarci = bow_vektorizer.get_feature_names_out()
bow_df = pd.DataFrame(bow_matris.toarray(), 
                       columns=kelime_dagarci,
                       index=[f'Belge {i+1}' for i in range(len(belgeler))])

print("Bag of Words Matrisi:")
print(bow_df.to_string())
print(f"\nKelime dağarcığı boyutu: {len(kelime_dagarci)}")
print(f"Matris şekli: {bow_matris.shape}")
```

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# En sık geçen kelimeler
kelime_frekanslari = bow_matris.toarray().sum(axis=0)
kelime_frekans_df = pd.DataFrame({'Kelime': kelime_dagarci, 'Frekans': kelime_frekanslari})
kelime_frekans_df = kelime_frekans_df.sort_values('Frekans', ascending=False).head(15)

plt.figure(figsize=(12, 5))
plt.bar(kelime_frekans_df['Kelime'], kelime_frekans_df['Frekans'], color='steelblue', edgecolor='black')
plt.title('En Sık Geçen Kelimeler (Bag of Words)', fontsize=14, fontweight='bold')
plt.xlabel('Kelime')
plt.ylabel('Frekans')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 6. TF-IDF (Term Frequency - Inverse Document Frequency)

TF-IDF, bir kelimenin bir belgede ne kadar önemli olduğunu ölçer:

- **TF (Term Frequency)**: Kelimenin belgede kaç kez geçtiği
- **IDF (Inverse Document Frequency)**: Kelimenin tüm belgelerde ne kadar nadir olduğu
- **TF-IDF = TF × IDF**: Sık geçen ama nadir olan kelimeler yüksek skor alır

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF vektörizasyonu
tfidf_vektorizer = TfidfVectorizer()
tfidf_matris = tfidf_vektorizer.fit_transform(belgeler)

# Sonuçları DataFrame olarak göster
tfidf_kelimeler = tfidf_vektorizer.get_feature_names_out()
tfidf_df = pd.DataFrame(tfidf_matris.toarray().round(3),
                         columns=tfidf_kelimeler,
                         index=[f'Belge {i+1}' for i in range(len(belgeler))])

print("TF-IDF Matrisi:")
print(tfidf_df.to_string())
print(f"\nMatris şekli: {tfidf_matris.shape}")
```

### Korelasyon Analizi ve Isı Haritası

Özellikler arasındaki ilişkiyi korelasyon matrisi ve ısı haritası ile görselleştiriyoruz. Yüksek korelasyonlu özellikler modelin en çok yararlanacağı özelliklerdir.

```python
# BoW ve TF-IDF karşılaştırması
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# BoW ısı haritası
import seaborn as sns
sns.heatmap(bow_df.iloc[:, :15], annot=True, cmap='YlOrRd', ax=axes[0], fmt='d')
axes[0].set_title('Bag of Words (İlk 15 Kelime)', fontsize=12, fontweight='bold')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=45, ha='right')

# TF-IDF ısı haritası
sns.heatmap(tfidf_df.iloc[:, :15], annot=True, cmap='YlOrRd', ax=axes[1], fmt='.2f')
axes[1].set_title('TF-IDF (İlk 15 Kelime)', fontsize=12, fontweight='bold')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45, ha='right')

plt.tight_layout()
plt.show()

print("\nFark: TF-IDF'te her belgede geçen kelimeler (ve, için, çok) düşük skor alır,")
print("nadir kelimeler ise yüksek skor alır.")
```

## 7. Kelime Bulutu (Word Cloud)

### Kütüphanelerin Yüklenmesi

Projede kullanacağımız kütüphaneleri içe aktarıyoruz:

| Kütüphane | Amacı |
|-----------|-------|
| `wordcloud` | Kelime bulutu görselleştirme |


```python
from wordcloud import WordCloud

# Türkçe metin derlem
turkce_derlem = """
Yapay zekâ teknolojisi hayatımızın her alanına girdi. Makine öğrenmesi ve derin öğrenme 
algoritmaları sayesinde bilgisayarlar artık görüntü tanıma, konuşma tanıma ve doğal dil 
işleme gibi karmaşık görevleri başarıyla yerine getirebiliyor. Büyük dil modelleri metin 
üretme, çeviri yapma ve soru cevaplama konularında insan düzeyinde performans sergiliyor.

Veri bilimi, yapay zekâ uygulamalarının temelini oluşturuyor. Büyük veri analizi, veri 
görselleştirme ve istatistiksel modelleme veri biliminin temel konularıdır. Python, veri 
bilimi ve yapay zekâ projelerinde en çok kullanılan programlama dilidir.

Derin öğrenme, özellikle evrişimli sinir ağları ve tekrarlayan sinir ağları alanlarında 
büyük başarılar elde etti. Transfer öğrenme sayesinde az veriyle bile yüksek performans 
elde edilebiliyor. Doğal dil işleme alanında transformer mimarisi devrim yarattı.

Yapay zekâ etiği ve güvenliği de önemli bir konu haline geldi. Yapay zekâ sistemlerinin 
adil, şeffaf ve hesap verebilir olması gerekiyor. Veri gizliliği ve önyargı konuları 
yapay zekâ araştırmacılarının gündeminde önemli bir yer tutuyor.
"""

# Durak kelimeleri çıkar
temiz_kelimeler = [k for k in turkce_derlem.lower().split() 
                   if k not in tum_durak and len(k) > 2 and k.isalpha()]
temiz_metin = ' '.join(temiz_kelimeler)

print(f"Toplam kelime: {len(turkce_derlem.split())}")
print(f"Temizleme sonrası: {len(temiz_kelimeler)}")
```

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Kelime bulutu oluştur
kelime_bulutu = WordCloud(
    width=800,
    height=400,
    background_color='white',
    colormap='viridis',
    max_words=100,
    max_font_size=100,
    random_state=42
).generate(temiz_metin)

plt.figure(figsize=(14, 7))
plt.imshow(kelime_bulutu, interpolation='bilinear')
plt.title('Yapay Zekâ Konulu Metin - Kelime Bulutu', fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Görselleştirme

Aşağıdaki grafikle veriyi görsel olarak inceliyoruz. Görselleştirme, sayısal analizlerin ötesinde örüntüleri ve anormallikleri fark etmemizi sağlar.

```python
# Kelime frekans analizi
from collections import Counter

kelime_sayaci = Counter(temiz_kelimeler)
en_sik_20 = kelime_sayaci.most_common(20)

kelimeler_list, frekanslar = zip(*en_sik_20)

plt.figure(figsize=(12, 6))
plt.barh(range(len(kelimeler_list)), frekanslar, color='steelblue', edgecolor='black')
plt.yticks(range(len(kelimeler_list)), kelimeler_list)
plt.xlabel('Frekans')
plt.title('En Sık Geçen 20 Kelime', fontsize=14, fontweight='bold')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
