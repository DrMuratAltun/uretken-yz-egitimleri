## 1. Kurulum

useanything.com → Mac/Win/Linux installer → kur (Ollama varsa otomatik bağlar).


## 2. İlk Workspace

"+ New Workspace" → adı ver (örn "Mesleki Bilgi Tabanım").

LLM seç:
- Ollama (lokal, ücretsiz, gizlilik max)
- OpenAI/Gemini/Claude (bulut, daha kaliteli ama API key gerekir)


## 3. Doküman Yükleme

Workspace → Documents → upload PDF/DOCX/TXT/MD/web URL.

AnythingLLM otomatik:
- Chunking (512 token, 64 overlap)
- Embedding (nomic-embed-text varsayılan)
- Vector DB (LanceDB lokal)


## 4. Soru-Cevap

Chat sekmesinde sor. AnythingLLM:
1. Sorunu embed eder
2. Vector DB'den en benzer 4 chunk'ı çeker
3. LLM'e bağlam + soru gönderir
4. Cevap + kaynak gösterir


## 5. NotebookLM'e Karşı

+ AnythingLLM: tamamen lokal, sınırsız, çoklu workspace.
+ NotebookLM: Audio Overview, Mind Map, daha güzel UI.

Karanlık dengesi: kişisel/iş için AnythingLLM, paylaşılan içerik üretimi için NotebookLM.
